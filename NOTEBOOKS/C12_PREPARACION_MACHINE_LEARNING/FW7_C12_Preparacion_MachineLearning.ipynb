# **FRAMEWORK V7 DATASET MAESTRO - PREPARACION PARA MACHINE LEARNING**

## **M0. Configuración**

```python
import pandas as pd

#==========================================================================================
# CARGA DEL DATASET MAESTRO
#==========================================================================================

url_dataset = (

"https://raw.githubusercontent.com/jriatiga/FRAMEWORK_V7/refs/heads/main/DATA/MASTER/C09_MASTER/Dataset_Maestro_Framework_v04.parquet"

)

dataset = pd.read_parquet(url_dataset)

print()

print("="*90)
print("DATASET MAESTRO")
print("="*90)

print()

print(f"Observaciones : {dataset.shape[0]}")

print(f"Variables     : {dataset.shape[1]}")
```

## **M1. Construcción del Dataset Optimizado para Machine Learning**

```python
#==========================================================================================
# VARIABLES SELECCIONADAS
#==========================================================================================

url_variables = (

"https://raw.githubusercontent.com/jriatiga/FRAMEWORK_V7/refs/heads/main/DATA/MASTER/C11_MASTER/Preparacion_Machine_Learning/ipm/variables_machine_learning.csv"

)

variables_ml = pd.read_csv(url_variables)

print()

print("="*90)
print("VARIABLES SELECCIONADAS")
print("="*90)

print()

print(f"Variables : {len(variables_ml)}")

display(variables_ml.head())
```

```python
#==========================================================================================
# LISTA DE VARIABLES
#==========================================================================================

lista_variables = variables_ml["Variable"].tolist()

print()

print("="*90)
print("VARIABLES UTILIZADAS")
print("="*90)

for variable in lista_variables:

    print(variable)
```

```python
#==========================================================================================
# VALIDACIÓN DE VARIABLES
#==========================================================================================

faltantes = [

    variable

    for variable in lista_variables

    if variable not in dataset.columns

]

print()

print("="*90)
print("VALIDACIÓN")
print("="*90)

print()

if len(faltantes)==0:

    print("Todas las variables fueron encontradas.")

else:

    print("Variables faltantes:")

    print(faltantes)
```

```python
#==========================================================================================
# CONSTRUCCIÓN DEL DATASET
#==========================================================================================

dataset_ml = dataset[lista_variables].copy()

print()

print("="*90)
print("DATASET OPTIMIZADO")
print("="*90)

print()

print(f"Observaciones : {dataset_ml.shape[0]}")

print(f"Variables     : {dataset_ml.shape[1]}")

display(dataset_ml.head())
```

```python
#==========================================================================================
# DIAGNÓSTICO GENERAL
#==========================================================================================

print()

print("="*90)
print("DIAGNÓSTICO")
print("="*90)

print()

print("Valores faltantes")

display(dataset_ml.isnull().sum())

print()

print("Duplicados")

print(dataset_ml.duplicated().sum())
```

```python
#==========================================================================================
# RESUMEN ESTADÍSTICO
#==========================================================================================

display(

    dataset_ml.describe().T

)
```

```python
#==========================================================================================
# EXPORTACIÓN
#==========================================================================================

import os
from pathlib import Path

ruta_str = "Preparacion_Machine_Learning/dataset_machine_learning/"

# Crear el directorio si no existe
ruta = Path(ruta_str)
ruta.mkdir(parents=True, exist_ok=True)

dataset_ml.to_parquet(

    ruta / "dataset_machine_learning.parquet",

    index=False

)

dataset_ml.to_csv(

    ruta / "dataset_machine_learning.csv",

    index=False,

    encoding="utf-8-sig"

)

dataset_ml.to_excel(

    ruta / "dataset_machine_learning.xlsx",

    index=False

)

print()

print("="*90)
print("ARCHIVOS GENERADOS")
print("="*90)

print()

print("dataset_machine_learning.parquet")

print("dataset_machine_learning.csv")

print("dataset_machine_learning.xlsx")
```
