# **FRAMEWORK V7 DATASET MAESTRO - PREPARACION PARA MACHINE LEARNING**

## **M0. Configuración**

```python
#==========================================================================================
# M0. CARGA DEL DATASET MAESTRO
#==========================================================================================

import pandas as pd

URL_DATASET = (
    "https://raw.githubusercontent.com/jriatiga/FRAMEWORK_V7/"
    "refs/heads/main/DATA/MASTER/C09_MASTER/"
    "Dataset_Maestro_Framework_v04.parquet"
)

dataset = pd.read_parquet(URL_DATASET)

print()
print("=" * 90)
print("DATASET MAESTRO - C12")
print("=" * 90)

print()
print(f"Observaciones     : {dataset.shape[0]}")
print(f"Variables         : {dataset.shape[1]}")
print(f"Duplicados        : {dataset.duplicated().sum()}")
print(f"Columnas únicas   : {dataset.columns.nunique()}")
print(f"Version_Dataset   : {'Version_Dataset' in dataset.columns}")

if "Version_Dataset" in dataset.columns:
    print()
    print("Versiones:")
    print(dataset["Version_Dataset"].value_counts(dropna=False))
```

## **M1. Construcción del Dataset Optimizado para Machine Learning**

```python
URL_VARIABLES_ML = (
    "https://raw.githubusercontent.com/jriatiga/FRAMEWORK_V7/"
    "refs/heads/main/DATA/MASTER/C11_MASTER/"
    "Preparacion_Machine_Learning/ipm/"
    "variables_machine_learning.csv"
)

variables_ml = pd.read_csv(
    URL_VARIABLES_ML,
    encoding="utf-8-sig"
)

print()
print("=" * 90)
print("VARIABLES SELECCIONADAS - C11")
print("=" * 90)

print()
print(f"Variables         : {len(variables_ml)}")
print(f"Variables únicas  : {variables_ml['Variable'].nunique()}")
print(f"Duplicadas        : {variables_ml['Variable'].duplicated().sum()}")
print(f"Nulos Variable    : {variables_ml['Variable'].isna().sum()}")

print()
print(
    variables_ml[
        ["Ranking", "Variable", "IPML", "Estado_Final"]
    ].to_string(index=False)
)
```

```python
#==========================================================================================
# LISTA DE VARIABLES PARA MACHINE LEARNING
#==========================================================================================

lista_variables = variables_ml["Variable"].tolist()

print()
print("=" * 90)
print("VARIABLES UTILIZADAS")
print("=" * 90)

print()
print(f"Total variables : {len(lista_variables)}")
print(f"Variables únicas: {len(set(lista_variables))}")

print()

for i, variable in enumerate(lista_variables, start=1):
    print(f"{i:02d}. {variable}")
```

```python
#==========================================================================================
# VALIDACIÓN DE VARIABLES EN EL DATASET MAESTRO
#==========================================================================================

faltantes = [
    variable
    for variable in lista_variables
    if variable not in dataset.columns
]

encontradas = [
    variable
    for variable in lista_variables
    if variable in dataset.columns
]

print()
print("=" * 90)
print("VALIDACIÓN DE VARIABLES")
print("=" * 90)

print()
print(f"Variables solicitadas : {len(lista_variables)}")
print(f"Variables encontradas : {len(encontradas)}")
print(f"Variables faltantes   : {len(faltantes)}")

print()

if len(faltantes) == 0:
    print("Todas las variables fueron encontradas correctamente.")
else:
    print("Variables faltantes:")
    for variable in faltantes:
        print("-", variable)
```

```python
#==========================================================================================
# CONSTRUCCIÓN DEL DATASET OPTIMIZADO
#==========================================================================================

dataset_ml = dataset[
    lista_variables
].copy()

print()
print("=" * 90)
print("DATASET OPTIMIZADO")
print("=" * 90)

print()
print(f"Observaciones : {dataset_ml.shape[0]}")
print(f"Variables     : {dataset_ml.shape[1]}")
print(f"Duplicados    : {dataset_ml.duplicated().sum()}")

print()
print("Columnas:")
for i, columna in enumerate(dataset_ml.columns, start=1):
    print(f"{i:02d}. {columna}")
```

```python
#==========================================================================================
# DIAGNÓSTICO GENERAL DEL DATASET OPTIMIZADO
#==========================================================================================

nulos_por_variable = dataset_ml.isna().sum()

print()
print("=" * 90)
print("DIAGNÓSTICO DEL DATASET OPTIMIZADO")
print("=" * 90)

print()
print("VALORES FALTANTES POR VARIABLE")
print("-" * 90)

print(nulos_por_variable)

print()
print("CONTROL")
print("-" * 90)

print("Variables con nulos :", (nulos_por_variable > 0).sum())
print("Total de nulos      :", nulos_por_variable.sum())
print("Duplicados          :", dataset_ml.duplicated().sum())
```

```python
#==========================================================================================
# RESUMEN ESTADÍSTICO
#==========================================================================================

resumen_estadistico = dataset_ml.describe().T

print()
print("=" * 90)
print("RESUMEN ESTADÍSTICO DEL DATASET OPTIMIZADO")
print("=" * 90)

print()
print(resumen_estadistico.to_string())

print()
print("CONTROL")
print("-" * 90)
print("Variables resumidas :", len(resumen_estadistico))
print("Count Volumen       :", resumen_estadistico.loc["VolumenUtilDiarioMasa", "count"])
print("Count IRCA          :", resumen_estadistico.loc["irca", "count"])
```

```python
#==========================================================================================
# EXPORTACIÓN DEL DATASET PARA MACHINE LEARNING
#==========================================================================================

from pathlib import Path
import pandas as pd

ruta = Path(
    "Preparacion_Machine_Learning/dataset_machine_learning"
)

ruta.mkdir(
    parents=True,
    exist_ok=True
)

#------------------------------------------------------------------------------------------
# Exportación
#------------------------------------------------------------------------------------------

dataset_ml.to_parquet(
    ruta / "dataset_machine_learning.parquet",
    index=False
)

dataset_ml.to_csv(
    ruta / "dataset_machine_learning.csv",
    index=False,
    encoding="utf-8-sig"
)

dataset_ml.to_excel(
    ruta / "dataset_machine_learning.xlsx",
    index=False
)

#------------------------------------------------------------------------------------------
# Control de archivos
#------------------------------------------------------------------------------------------

archivos = [
    "dataset_machine_learning.parquet",
    "dataset_machine_learning.csv",
    "dataset_machine_learning.xlsx"
]

print()
print("=" * 90)
print("EXPORTACIÓN DATASET MACHINE LEARNING")
print("=" * 90)

print()

for archivo in archivos:
    print(
        f"{archivo:<40}",
        f"existe={(ruta / archivo).exists()}"
    )

#------------------------------------------------------------------------------------------
# Relectura de control
#------------------------------------------------------------------------------------------

control_ml = pd.read_parquet(
    ruta / "dataset_machine_learning.parquet"
)

print()
print("CONTROL DEL ARCHIVO EXPORTADO")
print("-" * 90)

print("Dimensión          :", control_ml.shape)
print("Variables únicas   :", control_ml.columns.nunique())
print("Duplicados         :", control_ml.duplicated().sum())
print("Total nulos        :", control_ml.isna().sum().sum())
print(
    "Nulos Volumen     :",
    control_ml["VolumenUtilDiarioMasa"].isna().sum()
)

print(
    "Mismas columnas   :",
    control_ml.columns.tolist() == dataset_ml.columns.tolist()
)
```
