## **M3. Tratamiento de Valores Faltantes (Imputación y Consistencia)**

 ### **Clasificación de Variables Según Cobertura**

```python
print("="*70)
print("FASE 1 - CLASIFICACIÓN SEGÚN COBERTURA")
print("="*70)

cobertura = pd.DataFrame({
    "No_Nulos": dataset_maestro.notnull().sum(),
    "Nulos": dataset_maestro.isnull().sum()
})

cobertura["Cobertura_%"] = (
    cobertura["No_Nulos"] /
    len(dataset_maestro)
)*100

cobertura["Nulos_%"] = (
    cobertura["Nulos"] /
    len(dataset_maestro)
)*100

cobertura = cobertura.sort_values(
    "Cobertura_%",
    ascending=False
)

cobertura
```

```python
def clasificar(c):

    if c >= 95:
        return "Excelente"

    elif c >= 70:
        return "Alta"

    elif c >= 40:
        return "Media"

    elif c >= 15:
        return "Baja"

    else:
        return "Muy Baja"

cobertura["Categoria"] = cobertura["Cobertura_%"].apply(clasificar)

cobertura
```

```python
print("="*70)
print("RESUMEN POR CATEGORÍA")
print("="*70)

print(
    cobertura.groupby("Categoria").size()
)
```

```python
for categoria in [
    "Excelente",
    "Alta",
    "Media",
    "Baja",
    "Muy Baja"
]:

    print("\n")
    print("="*60)
    print(categoria)
    print("="*60)

    print(
        cobertura[
            cobertura["Categoria"]==categoria
        ].index.tolist()
    )
```

Esta fase no modifica ningún dato. Sirve para definir la estrategia de imputación que seguiremos en las siguientes fases.

| Categoría | Acción en el Paso 3 |
| :--- | :--- |
| **Excelente (≥95%)** | No requieren imputación. Se conservan tal como están. |
| **Alta (70–95%)** | Imputación simple (por ejemplo, propagación temporal o moda para variables categóricas). |
| **Media (40–70%)** | Imputación controlada utilizando la estructura temporal y espacial del dataset. |
| **Baja (15–40%)** | Imputación únicamente cuando exista suficiente contexto temporal; de lo contrario se conservan valores faltantes. |

#### **Imputación de VolumenUtilDiarioMasa**

**Etapa A: Reconstrucción de la Serie para 2025 (VolumenUtilDiarioMasa)**

Primero reconstruiremos febrero-diciembre de 2025, porque sólo falta el final de la serie. Usaremos una extrapolación basada en la estacionalidad del último año disponible (2024).

La idea es aplicar la siguiente relación para mantener la forma de la curva:

$$\text{Valor}_{2025} = \text{Valor}_{\text{Enero } 2025} \times \left( \frac{\text{Valor}_{\text{mes } 2024}}{\text{Valor}_{\text{Enero } 2024}} \right)$$

Esto permite proyectar los meses faltantes respetando el comportamiento estacional observado en el sistema hídrico compartido.

```python
import numpy as np

dataset_maestro["Volumen_Imputado"] = 0

for nodo in dataset_maestro["Nodo"].unique():

    datos = dataset_maestro[
        dataset_maestro["Nodo"] == nodo
    ]

    enero_2025 = datos[
        (datos["Anio"] == 2025) &
        (datos["Mes"] == 1)
    ]["VolumenUtilDiarioMasa"].iloc[0]

    enero_2024 = datos[
        (datos["Anio"] == 2024) &
        (datos["Mes"] == 1)
    ]["VolumenUtilDiarioMasa"].iloc[0]

    for mes in range(2,13):

        valor2024 = datos[
            (datos["Anio"] == 2024) &
            (datos["Mes"] == mes)
        ]["VolumenUtilDiarioMasa"].iloc[0]

        nuevo = enero_2025 * (valor2024 / enero_2024)

        mascara = (
            (dataset_maestro["Nodo"]==nodo) &
            (dataset_maestro["Anio"]==2025) &
            (dataset_maestro["Mes"]==mes)
        )

        dataset_maestro.loc[
            mascara,
            "VolumenUtilDiarioMasa"
        ] = nuevo

        dataset_maestro.loc[
            mascara,
            "Volumen_Imputado"
        ] = 1
```

```python
from scipy.interpolate import interp1d
import numpy as np

def imputar_por_interpolacion(dataset, variable, nombre_bandera=None):
    """
    Imputa valores faltantes por interpolación/extrapolación lineal
    utilizando la tendencia temporal de cada nodo.

    Parámetros
    ----------
    dataset : DataFrame
    variable : str
        Nombre de la variable a imputar.
    nombre_bandera : str (opcional)
        Nombre de la columna que indicará qué registros fueron imputados.
    """

    # Guardar los nulos originales
    nulos_originales = dataset[variable].isna()

    for nodo in dataset["Nodo"].unique():

        datos = dataset[dataset["Nodo"] == nodo]

        serie = (
            datos.groupby("Anio")[variable]
                 .first()
                 .sort_index()
        )

        # Años con información
        x = serie.dropna().index.values
        y = serie.dropna().values

        # Si no existen al menos dos datos, no se puede interpolar
        if len(x) < 2:
            continue

        f = interp1d(
            x,
            y,
            kind="linear",
            fill_value="extrapolate"
        )

        años = serie.index.values
        estimado = f(años)

        for año, valor in zip(años, estimado):

            mascara = (
                (dataset["Nodo"] == nodo) &
                (dataset["Anio"] == año)
            )

            dataset.loc[mascara, variable] = valor

    # Crear bandera de imputación
    if nombre_bandera is not None:
        dataset[nombre_bandera] = nulos_originales.astype(int)

    return dataset
```
