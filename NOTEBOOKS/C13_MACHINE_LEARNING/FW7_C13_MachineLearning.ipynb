## **M1. Diagnóstico Estadístico para Machine Learning**


```python
#==========================================================================================
# SCRIPT 26
# DIAGNÓSTICO ESTADÍSTICO PARA MACHINE LEARNING
#==========================================================================================
#
# Objetivo
#
# Evaluar el comportamiento estadístico de las variables
# seleccionadas para determinar la necesidad de aplicar
# transformaciones previas al entrenamiento.
#
#==========================================================================================
```

```python
print("="*90)
print("RESUMEN ESTADÍSTICO")
print("="*90)

display(dataset_ml.describe().T)
```

```python
print("="*90)
print("ASIMETRÍA")
print("="*90)

asimetria = pd.DataFrame({

    "Variable": dataset_ml.columns,

    "Asimetria": dataset_ml.skew().round(3)

})

display(asimetria)
```

```python
print("="*90)
print("CURTOSIS")
print("="*90)

curtosis = pd.DataFrame({

    "Variable": dataset_ml.columns,

    "Curtosis": dataset_ml.kurt().round(3)

})

display(curtosis)
```

```python
#==========================================================================================
# DETECCIÓN DE VALORES ATÍPICOS (IQR)
#==========================================================================================

outliers = []

for columna in dataset_ml.columns:

    # Variables binarias
    if dataset_ml[columna].nunique() <= 2:

        cantidad = np.nan

    else:

        q1 = dataset_ml[columna].quantile(0.25)

        q3 = dataset_ml[columna].quantile(0.75)

        iqr = q3 - q1

        inferior = q1 - 1.5 * iqr

        superior = q3 + 1.5 * iqr

        cantidad = (

            (

                dataset_ml[columna] < inferior

            )

            |

            (

                dataset_ml[columna] > superior

            )

        ).sum()

    outliers.append(

        [

            columna,

            cantidad

        ]

    )

outliers = pd.DataFrame(

    outliers,

    columns=[

        "Variable",

        "Outliers"

    ]

)

print()

print("="*90)
print("DETECCIÓN DE VALORES ATÍPICOS (IQR)")
print("="*90)

display(outliers)
```

```python
#==========================================================================================
# RECOMENDACIÓN PARA PREPROCESAMIENTO
#==========================================================================================

def recomendacion(variable, asimetria):

    # Variables binarias

    if dataset_ml[variable].nunique() <= 2:

        return "Variable binaria. No requiere transformación."

    if abs(asimetria) < 0.5:

        return "No requiere transformación."

    elif abs(asimetria) < 1:

        return "Evaluar estandarización."

    else:

        return "Considerar escalamiento robusto."
```

```python
diagnostico = asimetria.merge(

    outliers,

    on="Variable"

)

diagnostico["Recomendacion"] = diagnostico.apply(

    lambda fila:

    recomendacion(

        fila["Variable"],

        fila["Asimetria"]

    ),

    axis=1

)

display(diagnostico)
```

## **M2. Preparación, Transformación y Escalamiento del Dataset para Machine Learning - Experimentos**

Objetivo: Aplicar las transformaciones y técnicas de escalamiento definidas a partir del diagnóstico estadístico, generando un nuevo dataset optimizado para el entrenamiento de modelos de Machine Learning.

**Análisis de la Normalización y Optimización**

El proceso de normalización y optimización se realizó en varias etapas, buscando preparar el dataset de la mejor manera para el entrenamiento de modelos de Machine Learning. A continuación, se detalla el análisis de cada paso:

**1. Diagnóstico Estadístico**

Se realizó un diagnóstico exhaustivo de las variables del dataset para entender su distribución y detectar posibles problemas como asimetría, curtosis y valores atípicos. Los resultados clave fueron:

*   **Resumen Estadístico (`.describe().T`)**: Proporcionó una visión general de la centralidad, dispersión y forma de la distribución de cada variable.
*   **Asimetría (`.skew().round(3)`)**: Se calculó la asimetría para cada variable. Se identificaron variables con asimetría significativa (positiva o negativa), lo cual es crucial para la elección del método de escalado.
    *   `irca` y `VolumenUtilDiarioMasa` mostraron una fuerte asimetría positiva, debido a su naturaleza binaria.
    *   `Precipitacion_mm`, `ONI`, `Velocidad_Viento` mostraron ligera asimetría positiva.
    *   `Humedad_Relativa` mostró asimetría negativa.
*   **Curtosis (`.kurt().round(3)`)**: La curtosis confirmó la forma de las distribuciones, especialmente la concentración de valores en las variables binarias (`irca`, `VolumenUtilDiarioMasa`).
*   **Detección de Valores Atípicos (IQR)**: Se utilizó el método del rango intercuartílico (IQR) para identificar posibles valores atípicos. Se observó que este método identificaba 'outliers' en las variables binarias, lo cual se interpretó como una consecuencia de la naturaleza de estas variables y no como un problema real de datos atípicos, sugiriendo una mejora metodológica para excluir variables binarias de esta detección.

**2. Recomendación para Preprocesamiento**

Basado en el diagnóstico estadístico, se implementó una función (`recomendacion`) para sugerir el tipo de transformación necesario para cada variable:

*   **Variables binarias**: No requieren transformación.
*   **Asimetría `abs(asimetria) < 0.5`**: No requieren transformación.
*   **Asimetría `0.5 <= abs(asimetria) < 1`**: Evaluar estandarización.
*   **Asimetría `abs(asimetria) >= 1`**: Considerar escalamiento robusto (o transformación logarítmica si no es binaria).

Esta función generó un diagnóstico (`diagnostico`) que incluyó una columna de 'Recomendacion' para cada variable, guiando las decisiones de preprocesamiento.

**3. Aplicación Automática del Método de Transformación**

El Framework tomó una decisión automática sobre el método de escalado a aplicar a las variables predictoras (X) basándose en el número de variables con alta asimetría y/o presencia de outliers:

*   **Decisión**: Si el número de variables con asimetría `abs(asimetria) > 1` es igual o superior a 2, O si el número de variables con más de 10 outliers es igual o superior a 2, se opta por `RobustScaler`. En caso contrario, se selecciona `StandardScaler`.
*   **Ejecución**: Para el dataset actual, se determinó `RobustScaler` debido a la presencia de 3 variables muy asimétricas (`Precipitacion_mm`, `Humedad_Relativa`, `VolumenUtilDiarioMasa`) y 2 con outliers (`Precipitacion_mm`, `VolumenUtilDiarioMasa`).
*   **Resultado**: Las variables predictoras (`X`) fueron transformadas utilizando el `RobustScaler`, generando `X_transformado`. Esta transformación es más adecuada para datos que contienen muchos valores atípicos o no siguen una distribución normal, ya que utiliza la mediana y el rango intercuartílico, siendo menos sensible a los extremos.

Finalmente, el dataset transformado (`dataset_transformado`) fue reconstruido, uniendo las variables predictoras escaladas con la variable objetivo sin transformar. Este dataset está ahora optimizado y listo para la siguiente fase de construcción de secuencias temporales para el modelado de Machine Learning.

```python
#==========================================================================================
# CLASIFICACIÓN DE COBERTURA
#==========================================================================================

def clasificar_cobertura(x):

    if x >= 90:
        return "Excelente"

    elif x >= 70:
        return "Buena"

    elif x >= 50:
        return "Aceptable"

    elif x >= 20:
        return "Baja"

    else:
        return "Insuficiente"

cobertura["Nivel_Cobertura"] = (

    cobertura["% Cobertura"]

    .apply(clasificar_cobertura)

)

display(cobertura)
```

```python
#==========================================================================================
# DECISIÓN DEL FRAMEWORK
#==========================================================================================

UMBRAL_COBERTURA = 50

cobertura["Decision"] = "Excluir"

cobertura.loc[
    cobertura["% Cobertura"] >= UMBRAL_COBERTURA,
    "Decision"
] = "Utilizar"

print()

print("="*90)
print("DECISIÓN DEL FRAMEWORK")
print("="*90)

print()

display(cobertura)
```
