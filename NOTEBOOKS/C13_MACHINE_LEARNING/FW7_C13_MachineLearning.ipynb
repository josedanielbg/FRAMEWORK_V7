```python
#==========================================================================================
# SCRIPT 26
# DIAGNÓSTICO ESTADÍSTICO PARA MACHINE LEARNING
#==========================================================================================
#
# Objetivo
#
# Evaluar el comportamiento estadístico de las variables
# seleccionadas para determinar la necesidad de aplicar
# transformaciones previas al entrenamiento.
#
#==========================================================================================
```

```python
print("="*90)
print("RESUMEN ESTADÍSTICO")
print("="*90)

display(dataset_ml.describe().T)
```

```python
print("="*90)
print("ASIMETRÍA")
print("="*90)

asimetria = pd.DataFrame({

    "Variable": dataset_ml.columns,

    "Asimetria": dataset_ml.skew().round(3)

})

display(asimetria)
```

```python
print("="*90)
print("CURTOSIS")
print("="*90)

curtosis = pd.DataFrame({

    "Variable": dataset_ml.columns,

    "Curtosis": dataset_ml.kurt().round(3)

})

display(curtosis)
```

```python
#==========================================================================================
# DETECCIÓN DE VALORES ATÍPICOS (IQR)
#==========================================================================================

outliers = []

for columna in dataset_ml.columns:

    # Variables binarias
    if dataset_ml[columna].nunique() <= 2:

        cantidad = np.nan

    else:

        q1 = dataset_ml[columna].quantile(0.25)

        q3 = dataset_ml[columna].quantile(0.75)

        iqr = q3 - q1

        inferior = q1 - 1.5 * iqr

        superior = q3 + 1.5 * iqr

        cantidad = (

            (

                dataset_ml[columna] < inferior

            )

            |

            (

                dataset_ml[columna] > superior

            )

        ).sum()

    outliers.append(

        [

            columna,

            cantidad

        ]

    )

outliers = pd.DataFrame(

    outliers,

    columns=[

        "Variable",

        "Outliers"

    ]

)

print()

print("="*90)
print("DETECCIÓN DE VALORES ATÍPICOS (IQR)")
print("="*90)

display(outliers)
```

```python
variable = "VolumenUtilDiarioMasa"

print("Mínimo:", dataset_ml[variable].min())
print("Máximo:", dataset_ml[variable].max())
print("Media:", dataset_ml[variable].mean())

print("\nValores únicos:")
print(
    sorted(
        dataset_ml[variable].unique()
    )
)

print("\nFrecuencias:")
print(
    dataset_ml[variable].value_counts(
        dropna=False
    ).sort_index()
)
```

```python
#==========================================================================================
# RECOMENDACIÓN PARA PREPROCESAMIENTO
#==========================================================================================

def recomendacion(variable, asimetria):

    # Variables binarias

    if dataset_ml[variable].nunique() <= 2:

        return "Variable binaria. No requiere transformación."

    if abs(asimetria) < 0.5:

        return "No requiere transformación."

    elif abs(asimetria) < 1:

        return "Evaluar estandarización."

    else:

        return "Considerar escalamiento robusto."
```

```python
diagnostico = asimetria.merge(

    outliers,

    on="Variable"

)

diagnostico["Recomendacion"] = diagnostico.apply(

    lambda fila:

    recomendacion(

        fila["Variable"],

        fila["Asimetria"]

    ),

    axis=1

)

display(diagnostico)
```

## **M2. Preparación, Transformación y Escalamiento del Dataset para Machine Learning - Experimentos**

Objetivo: Aplicar las transformaciones y técnicas de escalamiento definidas a partir del diagnóstico estadístico, generando un nuevo dataset optimizado para el entrenamiento de modelos de Machine Learning.

**Análisis de la Normalización y Optimización**

El proceso de normalización y optimización se realizó en varias etapas, buscando preparar el dataset de la mejor manera para el entrenamiento de modelos de Machine Learning. A continuación, se detalla el análisis de cada paso:

**1. Diagnóstico Estadístico**

Se realizó un diagnóstico exhaustivo de las variables del dataset para entender su distribución y detectar posibles problemas como asimetría, curtosis y valores atípicos. Los resultados clave fueron:

*   **Resumen Estadístico (`.describe().T`)**: Proporcionó una visión general de la centralidad, dispersión y forma de la distribución de cada variable.
*   **Asimetría (`.skew().round(3)`)**: Se calculó la asimetría para cada variable. Se identificaron variables con asimetría significativa (positiva o negativa), lo cual es crucial para la elección del método de escalado.
    *   `irca` y `VolumenUtilDiarioMasa` mostraron una fuerte asimetría positiva, debido a su naturaleza binaria.
    *   `Precipitacion_mm`, `ONI`, `Velocidad_Viento` mostraron ligera asimetría positiva.
    *   `Humedad_Relativa` mostró asimetría negativa.
*   **Curtosis (`.kurt().round(3)`)**: La curtosis confirmó la forma de las distribuciones, especialmente la concentración de valores en las variables binarias (`irca`, `VolumenUtilDiarioMasa`).
*   **Detección de Valores Atípicos (IQR)**: Se utilizó el método del rango intercuartílico (IQR) para identificar posibles valores atípicos. Se observó que este método identificaba 'outliers' en las variables binarias, lo cual se interpretó como una consecuencia de la naturaleza de estas variables y no como un problema real de datos atípicos, sugiriendo una mejora metodológica para excluir variables binarias de esta detección.

**2. Recomendación para Preprocesamiento**

Basado en el diagnóstico estadístico, se implementó una función (`recomendacion`) para sugerir el tipo de transformación necesario para cada variable:

*   **Variables binarias**: No requieren transformación.
*   **Asimetría `abs(asimetria) < 0.5`**: No requieren transformación.
*   **Asimetría `0.5 <= abs(asimetria) < 1`**: Evaluar estandarización.
*   **Asimetría `abs(asimetria) >= 1`**: Considerar escalamiento robusto (o transformación logarítmica si no es binaria).

Esta función generó un diagnóstico (`diagnostico`) que incluyó una columna de 'Recomendacion' para cada variable, guiando las decisiones de preprocesamiento.

**3. Aplicación Automática del Método de Transformación**

El Framework tomó una decisión automática sobre el método de escalado a aplicar a las variables predictoras (X) basándose en el número de variables con alta asimetría y/o presencia de outliers:

*   **Decisión**: Si el número de variables con asimetría `abs(asimetria) > 1` es igual o superior a 2, O si el número de variables con más de 10 outliers es igual o superior a 2, se opta por `RobustScaler`. En caso contrario, se selecciona `StandardScaler`.
*   **Ejecución**: Para el dataset actual, se determinó `RobustScaler` debido a la presencia de 3 variables muy asimétricas (`Precipitacion_mm`, `Humedad_Relativa`, `VolumenUtilDiarioMasa`) y 2 con outliers (`Precipitacion_mm`, `VolumenUtilDiarioMasa`).
*   **Resultado**: Las variables predictoras (`X`) fueron transformadas utilizando el `RobustScaler`, generando `X_transformado`. Esta transformación es más adecuada para datos que contienen muchos valores atípicos o no siguen una distribución normal, ya que utiliza la mediana y el rango intercuartílico, siendo menos sensible a los extremos.

Finalmente, el dataset transformado (`dataset_transformado`) fue reconstruido, uniendo las variables predictoras escaladas con la variable objetivo sin transformar. Este dataset está ahora optimizado y listo para la siguiente fase de construcción de secuencias temporales para el modelado de Machine Learning.

```python
#==========================================================================================
# CODIFICACIÓN DE LA VARIABLE OBJETIVO DEL MODELO
#==========================================================================================

MAPEO_OBJETIVO = None

if TIPO_PROBLEMA == "Clasificación":

    if VARIABLE_OBJETIVO_MODELO == "nivel_de_riesgo":

        MAPEO_OBJETIVO = {
            "Sin riesgo": 0,
            "Riesgo bajo": 1
        }

        clases_observadas = set(
            y.dropna().unique()
        )

        clases_esperadas = set(
            MAPEO_OBJETIVO.keys()
        )

        clases_no_reconocidas = (
            clases_observadas
            - clases_esperadas
        )

        if clases_no_reconocidas:

            raise Exception(
                "Se encontraron categorías no definidas "
                f"en el mapeo: {clases_no_reconocidas}"
            )

        y_original = y.copy()

        y = (
            y.map(MAPEO_OBJETIVO)
            .astype("int8")
        )

    else:

        raise Exception(
            "No existe un mapeo explícito definido para "
            f"la variable objetivo '{VARIABLE_OBJETIVO_MODELO}'."
        )

else:

    # En regresión el objetivo permanece numérico
    y_original = y.copy()


print()

print("=" * 90)
print("CODIFICACIÓN DE LA VARIABLE OBJETIVO")
print("=" * 90)

print()

print(
    f"Tipo de problema     : {TIPO_PROBLEMA}"
)

print(
    f"Variable objetivo    : {VARIABLE_OBJETIVO_MODELO}"
)

print(
    f"Mapeo aplicado       : {MAPEO_OBJETIVO}"
)

print(
    f"Tipo final de y      : {y.dtype}"
)

print()

print("Distribución final de y:")

print(
    y.value_counts(
        dropna=False
    ).sort_index()
)
```

```python
#==========================================================================================
# DIAGNÓSTICO ESTADÍSTICO DE VARIABLES PREDICTORAS
#==========================================================================================

import numpy as np
import pandas as pd

#------------------------------------------------------------------------------------------
# 1. ASIMETRÍA
#------------------------------------------------------------------------------------------

asimetria = pd.DataFrame({
    "Variable": X.columns,
    "Asimetria": X.skew().round(3)
}).reset_index(drop=True)

#------------------------------------------------------------------------------------------
# 2. DETECCIÓN DE VALORES ATÍPICOS (IQR)
#------------------------------------------------------------------------------------------

outliers = []

for columna in X.columns:

    # Variables binarias
    if X[columna].nunique() <= 2:

        cantidad = np.nan

    else:

        q1 = X[columna].quantile(0.25)
        q3 = X[columna].quantile(0.75)

        iqr = q3 - q1

        inferior = q1 - 1.5 * iqr
        superior = q3 + 1.5 * iqr

        cantidad = (
            (X[columna] < inferior)
            |
            (X[columna] > superior)
        ).sum()

    outliers.append(
        [
            columna,
            cantidad
        ]
    )

outliers = pd.DataFrame(
    outliers,
    columns=[
        "Variable",
        "Outliers"
    ]
)

#------------------------------------------------------------------------------------------
# 3. RECOMENDACIÓN DE PREPROCESAMIENTO
#------------------------------------------------------------------------------------------

def recomendacion(variable, asimetria):

    if X[variable].nunique() <= 2:

        return "Variable binaria. No requiere transformación."

    if abs(asimetria) < 0.5:

        return "No requiere transformación."

    elif abs(asimetria) < 1:

        return "Evaluar estandarización."

    else:

        return "Considerar escalamiento robusto."


diagnostico = asimetria.merge(
    outliers,
    on="Variable"
)

diagnostico["Recomendacion"] = diagnostico.apply(
    lambda fila: recomendacion(
        fila["Variable"],
        fila["Asimetria"]
    ),
    axis=1
)

#------------------------------------------------------------------------------------------
# CONTROLES
#------------------------------------------------------------------------------------------

print()

print("=" * 90)
print("DIAGNÓSTICO ESTADÍSTICO")
print("=" * 90)

print()

print(
    f"Variables predictoras diagnosticadas : {len(diagnostico)}"
)

print(
    f"Variables esperadas                  : {X.shape[1]}"
)

print(
    f"Objetivo científico incluido         : {VARIABLE_OBJETIVO in diagnostico['Variable'].values}"
)

print(
    f"Objetivo modelo incluido              : {VARIABLE_OBJETIVO_MODELO in diagnostico['Variable'].values}"
)

print()

display(diagnostico)
```

```python
#==========================================================================================
# SELECCIÓN AUTOMÁTICA DEL MÉTODO DE TRANSFORMACIÓN
#==========================================================================================

variables_asimetricas = diagnostico.loc[
    diagnostico["Asimetria"].abs() > 1
].shape[0]

variables_outliers = diagnostico.loc[
    diagnostico["Outliers"] > 10
].shape[0]

if (
    variables_asimetricas >= 2
    or
    variables_outliers >= 2
):

    metodo_transformacion = "RobustScaler"

else:

    metodo_transformacion = "StandardScaler"

print()

print("="*90)
print("DECISIÓN DEL FRAMEWORK")
print("="*90)

print()

print(f"Variables muy asimétricas : {variables_asimetricas}")
print(f"Variables con outliers    : {variables_outliers}")

print()

print(f"Método seleccionado : {metodo_transformacion}")
```

```python
#==========================================================================================
# CONSTRUCCIÓN DE SECUENCIAS TEMPORALES POR NODO
#==========================================================================================

import numpy as np
import pandas as pd

#------------------------------------------------------------------------------------------
# Variables predictoras y objetivo
#------------------------------------------------------------------------------------------

X = dataset_transformado.drop(
    columns=[VARIABLE_OBJETIVO_MODELO]
).copy()

y = dataset_transformado[
    VARIABLE_OBJETIVO_MODELO
].copy()

variables_predictoras_secuencia = X.columns.tolist()

#------------------------------------------------------------------------------------------
# Integración temporal
#
# Nodo y Fecha se utilizan exclusivamente para ordenar y agrupar.
# NO forman parte de las variables predictoras.
#------------------------------------------------------------------------------------------

datos_secuencias = pd.concat(
    [
        metadata_temporal[
            ["Nodo", "Fecha"]
        ].reset_index(drop=True),

        dataset_transformado.reset_index(drop=True)
    ],
    axis=1
)

datos_secuencias["Fecha"] = pd.to_datetime(
    datos_secuencias["Fecha"]
)

#------------------------------------------------------------------------------------------
# Contenedores
#------------------------------------------------------------------------------------------

secuencias_X = []
secuencias_y = []

metadata_secuencias_detalle = []

secuencias_potenciales = 0
secuencias_omitidas = 0

resumen_por_nodo = []

#------------------------------------------------------------------------------------------
# Construcción independiente por Nodo
#------------------------------------------------------------------------------------------

for nodo, grupo in datos_secuencias.groupby(
    "Nodo",
    sort=False
):

    grupo = grupo.sort_values(
        "Fecha"
    ).reset_index(drop=True)

    X_nodo = grupo[
        variables_predictoras_secuencia
    ]

    y_nodo = grupo[
        VARIABLE_OBJETIVO_MODELO
    ]

    potenciales_nodo = (
        len(grupo)
        - VENTANA
        - HORIZONTE
        + 1
    )

    potenciales_nodo = max(
        0,
        potenciales_nodo
    )

    validas_nodo = 0
    omitidas_nodo = 0

    secuencias_potenciales += potenciales_nodo

    for i in range(
        potenciales_nodo
    ):

        indice_objetivo = (
            i
            + VENTANA
            + HORIZONTE
            - 1
        )

        ventana_X = X_nodo.iloc[
            i:i + VENTANA
        ]

        objetivo = y_nodo.iloc[
            indice_objetivo
        ]

        #----------------------------------------------------------------------
        # Omitir únicamente secuencias incompletas
        #----------------------------------------------------------------------

        if (
            ventana_X.isna().any().any()
            or
            pd.isna(objetivo)
        ):

            secuencias_omitidas += 1
            omitidas_nodo += 1

            continue

        secuencias_X.append(
            ventana_X.to_numpy()
        )

        secuencias_y.append(
            objetivo
        )

        metadata_secuencias_detalle.append(
            {
                "Nodo": nodo,
                "Fecha_Inicio": grupo.loc[
                    i,
                    "Fecha"
                ],
                "Fecha_Fin": grupo.loc[
                    i + VENTANA - 1,
                    "Fecha"
                ],
                "Fecha_Objetivo": grupo.loc[
                    indice_objetivo,
                    "Fecha"
                ]
            }
        )

        validas_nodo += 1

    resumen_por_nodo.append(
        {
            "Nodo": nodo,
            "Observaciones": len(grupo),
            "Secuencias_Potenciales": potenciales_nodo,
            "Secuencias_Omitidas": omitidas_nodo,
            "Secuencias_Validas": validas_nodo
        }
    )

#------------------------------------------------------------------------------------------
# Metadatos de las secuencias
#------------------------------------------------------------------------------------------

metadata_secuencias_detalle = pd.DataFrame(
    metadata_secuencias_detalle
)

resumen_por_nodo = pd.DataFrame(
    resumen_por_nodo
)

#------------------------------------------------------------------------------------------
# Controles
#------------------------------------------------------------------------------------------

print()

print("=" * 90)
print("SECUENCIAS TEMPORALES POR NODO")
print("=" * 90)

print()

print(
    f"Variable objetivo       : {VARIABLE_OBJETIVO_MODELO}"
)

print(
    f"Variables predictoras   : {X.shape[1]}"
)

print(
    f"Nodos procesados        : {resumen_por_nodo.shape[0]}"
)

print(
    f"Secuencias potenciales  : {secuencias_potenciales}"
)

print(
    f"Secuencias omitidas     : {secuencias_omitidas}"
)

print(
    f"Secuencias válidas      : {len(secuencias_X)}"
)

print()

print("Distribución del objetivo:")

print(
    pd.Series(
        secuencias_y,
        name=VARIABLE_OBJETIVO_MODELO
    ).value_counts().sort_index()
)

print()

display(
    resumen_por_nodo
)
```

```python
#==========================================================================================
# INTEGRIDAD TEMPORAL DE LAS SECUENCIAS
#==========================================================================================

import numpy as np
import pandas as pd

#------------------------------------------------------------------------------------------
# Recalcular cuántas secuencias son realmente esperables por Nodo
#
# Una secuencia será válida solamente cuando:
#
# 1. exista la ventana temporal completa;
# 2. todas las predictoras de la ventana tengan datos;
# 3. exista el valor objetivo en el horizonte correspondiente.
#
# Esto permite trabajar correctamente tanto con:
# - objetivos con cobertura completa, como Exp01;
# - objetivos con cobertura parcial, como Exp04.
#------------------------------------------------------------------------------------------

control_nodos = []

for nodo, grupo in datos_secuencias.groupby(
    "Nodo",
    sort=False
):

    grupo = (
        grupo
        .sort_values("Fecha")
        .reset_index(drop=True)
    )

    observaciones_nodo = len(
        grupo
    )

    secuencias_potenciales = max(
        0,
        observaciones_nodo
        - VENTANA
        - HORIZONTE
        + 1
    )

    secuencias_validas_esperadas = 0

    #--------------------------------------------------------------------------
    # Evaluar cada ventana potencial
    #--------------------------------------------------------------------------

    for i in range(
        secuencias_potenciales
    ):

        indice_objetivo = (
            i
            + VENTANA
            + HORIZONTE
            - 1
        )

        ventana_x = grupo.iloc[
            i:i + VENTANA
        ][
            X.columns
        ]

        objetivo = grupo.iloc[
            indice_objetivo
        ][
            VARIABLE_OBJETIVO_MODELO
        ]

        ventana_valida = (
            not ventana_x
            .isna()
            .any()
            .any()
        )

        objetivo_valido = (
            pd.notna(objetivo)
        )

        if (
            ventana_valida
            and objetivo_valido
        ):

            secuencias_validas_esperadas += 1

    #--------------------------------------------------------------------------
    # Secuencias realmente generadas para el Nodo
    #--------------------------------------------------------------------------

    secuencias_obtenidas = int(
        (
            metadata_secuencias_detalle[
                "Nodo"
            ]
            == nodo
        ).sum()
    )

    control_nodos.append({

        "Nodo":
            nodo,

        "Observaciones":
            observaciones_nodo,

        "Potenciales":
            secuencias_potenciales,

        "Esperadas_Validas":
            secuencias_validas_esperadas,

        "Obtenidas":
            secuencias_obtenidas,

        "Conteo_Correcto":
            (
                secuencias_validas_esperadas
                ==
                secuencias_obtenidas
            )

    })

#------------------------------------------------------------------------------------------
# Tabla de control
#------------------------------------------------------------------------------------------

control_nodos = pd.DataFrame(
    control_nodos
)

#------------------------------------------------------------------------------------------
# Controles globales
#------------------------------------------------------------------------------------------

alineacion_global = (
    len(X_array)
    ==
    len(y_array)
    ==
    len(metadata_secuencias_detalle)
)

fechas_completas = (
    metadata_secuencias_detalle[
        [
            "Nodo",
            "Fecha_Inicio",
            "Fecha_Fin",
            "Fecha_Objetivo"
        ]
    ]
    .notna()
    .all()
    .all()
)

orden_temporal_correcto = (

    (
        metadata_secuencias_detalle[
            "Fecha_Inicio"
        ]
        <=
        metadata_secuencias_detalle[
            "Fecha_Fin"
        ]
    )
    &
    (
        metadata_secuencias_detalle[
            "Fecha_Fin"
        ]
        <
        metadata_secuencias_detalle[
            "Fecha_Objetivo"
        ]
    )

).all()

nodos_fuente = set(
    datos_secuencias[
        "Nodo"
    ].unique()
)

nodos_secuencias = set(
    metadata_secuencias_detalle[
        "Nodo"
    ].unique()
)

nodos_coinciden = (
    nodos_fuente
    ==
    nodos_secuencias
)

conteo_correcto_por_nodo = (
    control_nodos[
        "Conteo_Correcto"
    ]
    .all()
)

#------------------------------------------------------------------------------------------
# Índice secuencial, si ya fue incorporado durante la exportación
#------------------------------------------------------------------------------------------

if (
    "Indice_Secuencia"
    in metadata_secuencias_detalle.columns
):

    indices_secuenciales = np.array_equal(

        metadata_secuencias_detalle[
            "Indice_Secuencia"
        ].to_numpy(),

        np.arange(
            len(metadata_secuencias_detalle)
        )

    )

else:

    indices_secuenciales = True

#------------------------------------------------------------------------------------------
# Integridad global
#------------------------------------------------------------------------------------------

integridad_global = all([

    alineacion_global,
    fechas_completas,
    orden_temporal_correcto,
    nodos_coinciden,
    conteo_correcto_por_nodo,
    indices_secuenciales

])

#------------------------------------------------------------------------------------------
# Reporte
#------------------------------------------------------------------------------------------

print()

print("=" * 90)
print("INTEGRIDAD TEMPORAL DE LAS SECUENCIAS")
print("=" * 90)

print()

print(
    f"Secuencias en X             : {len(X_array)}"
)

print(
    f"Objetivos en y              : {len(y_array)}"
)

print(
    f"Registros metadata          : {len(metadata_secuencias_detalle)}"
)

print(
    f"Índices secuenciales        : {indices_secuenciales}"
)

print(
    f"Fechas completas            : {fechas_completas}"
)

print(
    f"Orden temporal correcto     : {orden_temporal_correcto}"
)

print(
    f"Nodos coinciden             : {nodos_coinciden}"
)

print(
    f"Conteo correcto por Nodo    : {conteo_correcto_por_nodo}"
)

print()

display(
    control_nodos
)

print()

print(
    f"INTEGRIDAD GLOBAL           : {integridad_global}"
)
```
