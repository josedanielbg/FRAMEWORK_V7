---

**Construcción del IPML**

Una vez consolidadas las cuatro dimensiones en el Universo Analítico, se define la ecuación oficial del **Índice de Priorización para Machine Learning (IPML)**.

---

**Ecuación Definitiva**

$$IPML = 0.30 \times PCD + 0.30 \times PC + 0.20 \times PR + 0.20 \times PM$$

**Donde:**
* **$PCD$**: Puntaje de Conocimiento del Dominio
* **$PC$**: Puntaje de Calidad
* **$PR$**: Puntaje de Redundancia
* **$PM$**: Puntaje de Multicolinealidad

---

**Justificación Técnica de la Ponderación**

Los pesos asignados responden a la estructura y principios del Framework sustentado en la investigación:

| Dimensión | Peso | Justificación |
| :--- | :---: | :--- |
| **Conocimiento del Dominio ($PCD$)** | **30%** | Es la dimensión transversal más importante; representa el criterio experto sobre la relevancia temática y técnica de la variable para la gestión hídrica. |
| **Calidad de Datos ($PC$)** | **30%** | Evalúa la integridad de la variable. Una variable relevante pierde su valor práctico e introduce sesgo si sus datos no son fiables o están incompletos. |
| **Redundancia ($PR$)** | **20%** | Minimiza la duplicidad de información directa entre pares de variables, optimizando la dimensionalidad del conjunto de datos. |
| **Multicolinealidad ($PM$)** | **20%** | Favorece la estabilidad matemática de los modelos finales reduciendo dependencias lineales complejas entre el grupo de variables. |

## **M3. Calculo Índice de Priorización para Machine Learning (IPML)**

Integrar las cuatro dimensiones obtenidas en los módulos anteriores para calcular el Índice de Priorización para Machine Learning (IPML) y generar el ranking final de variables.

```python
#==========================================================================================
# RECONSTRUCCIÓN DEL PUNTAJE DE MULTICOLINEALIDAD
#==========================================================================================

mapa_vif = {

    "Perfecta":0.00,

    "Muy Alta":0.25,

    "Alta":0.50,

    "Moderada":0.75,

    "Baja":1.00,

    "Sin Multicolinealidad":1.00,

    "No Calculado":1.00

}

universo["Puntaje_Multicolinealidad"] = (

    universo["Estado_VIF"]

    .map(mapa_vif)

)

display(

    universo[
        [

            "Variable",

            "Estado_VIF",

            "Puntaje_Multicolinealidad"

        ]

    ].head()

)
```

```python
#==========================================================================================
# PESOS DEL IPML
#==========================================================================================

peso_dominio = 0.40

peso_redundancia = 0.30

peso_multicolinealidad = 0.20

peso_calidad = 0.10

print()

print("="*90)
print("PESOS DEL IPML")
print("="*90)

print()

print(f"Conocimiento del Dominio : {peso_dominio:.2f}")
print(f"Redundancia              : {peso_redundancia:.2f}")
print(f"Multicolinealidad        : {peso_multicolinealidad:.2f}")
print(f"Calidad                  : {peso_calidad:.2f}")

print()

print(f"Suma de pesos            : {peso_dominio + peso_redundancia + peso_multicolinealidad + peso_calidad:.2f}")
```

```python
#==========================================================================================
# CÁLCULO DEL ÍNDICE DE PRIORIZACIÓN PARA MACHINE LEARNING (IPML)
#==========================================================================================

universo["IPML"] = (

      peso_dominio *

      universo["Puntaje_Conocimiento_Dominio"]

    +

      peso_calidad *

      universo["Puntaje_Calidad"]

    +

      peso_redundancia *

      universo["Puntaje_Redundancia"]

    +

      peso_multicolinealidad *

      universo["Puntaje_Multicolinealidad"]

).round(4)

print()

print("="*90)
print("ÍNDICE DE PRIORIZACIÓN PARA MACHINE LEARNING")
print("="*90)

display(

    universo[

        [

            "Variable",

            "IPML"

        ]

    ].head(10)

)
```

```python
#==========================================================================================
# RANKING DEL IPML
#==========================================================================================

ranking_ipml = (

    universo

    .sort_values(

        "IPML",

        ascending=False

    )

    .reset_index(drop=True)

)

ranking_ipml.insert(

    0,

    "Ranking",

    ranking_ipml.index + 1

)

print()

print("="*90)
print("RANKING DEL IPML")
print("="*90)

display(

    ranking_ipml.head(15)

)
```

```python
#==========================================================================================
# CLASIFICACIÓN DEL IPML
#==========================================================================================

def clasificar_ipml(valor):

    if valor >= 0.90:
        return "Muy Alta"

    elif valor >= 0.80:
        return "Alta"

    elif valor >= 0.70:
        return "Media"

    else:
        return "Baja"


universo["Nivel_IPML"] = universo["IPML"].apply(clasificar_ipml)

print()

print("="*90)
print("CLASIFICACIÓN DEL IPML")
print("="*90)

display(

    universo[
        [
            "Variable",
            "IPML",
            "Nivel_IPML"
        ]
    ].sort_values(
        "IPML",
        ascending=False
    ).head(15)

)
```

## **M4. Selección del Universo Definitivo para Machine Learning**

```python
#==========================================================================================
# SCRIPT
# SELECCIÓN DEL UNIVERSO DEFINITIVO PARA MACHINE LEARNING
#==========================================================================================
#
# Objetivo:
#
# Seleccionar las variables que conformarán el Dataset definitivo para
# Machine Learning utilizando el Índice de Priorización (IPML) y criterios
# adicionales de calidad y cobertura.
#
# Producto esperado:
#
# • Dataset de variables seleccionadas
# • Ranking final de variables
#
#==========================================================================================
```

```python
import os
import pandas as pd

# Definir las rutas
url = "https://raw.githubusercontent.com/jriatiga/FRAMEWORK_V7/refs/heads/main/DATA/MASTER/C11_MASTER/Universo_Analitico/ranking_ipml_integracion_dimensiones.csv"
ruta_local_ranking = "ranking_ipml_integracion_dimensiones.csv"  # Ajusta la ruta si lo tienes en otra carpeta

try:
    # Intentar cargar desde GitHub
    print("Intentando cargar el ranking IPML desde GitHub...")
    ranking = pd.read_csv(url)
    print("¡Carga exitosa desde GitHub!")
except Exception as e:
    print(f"No se pudo cargar desde GitHub ({e}). Intentando cargar desde el archivo local...")
    # Cargar desde la ruta local si falla la web
    ranking = pd.read_csv(ruta_local_ranking)
    print("¡Carga exitosa desde el archivo local!")

print()
print("=" * 90)
print("RANKING IPML")
print("=" * 90)
print()

print(f"Variables : {len(ranking)}")

display(ranking.head())
```

```python
#==========================================================================================
# CRITERIOS DE SELECCIÓN
#==========================================================================================

COBERTURA_MINIMA = 30

IPML_MINIMO = 0.80

print()

print("="*90)
print("CRITERIOS DEL FRAMEWORK")
print("="*90)

print()

print(f"IPML mínimo              : {IPML_MINIMO}")

print(f"Cobertura mínima (%)     : {COBERTURA_MINIMA}")

print("Estado VIF permitido     : diferente de 'Perfecta'")
```

```python
#==========================================================================================
# ESTADO FINAL DE LAS VARIABLES
#==========================================================================================

def clasificar_variable(fila):

    #----------------------------------------------------------
    # Validar IPML no calculado
    #----------------------------------------------------------

    if pd.isna(fila["IPML"]):
        return "No Evaluada"

    #----------------------------------------------------------

    if fila["IPML"] < IPML_MINIMO:
        return "No Prioritaria"

    if fila["% Cobertura"] < COBERTURA_MINIMA:
        return "Pendiente por Cobertura"

    if fila["Estado_VIF"] == "Perfecta":
        return "Revisar por Multicolinealidad"

    return "Seleccionada"


ranking["Estado_Final"] = ranking.apply(

    clasificar_variable,

    axis=1

)

display(

    ranking[
        [
            "Variable",
            "IPML",
            "% Cobertura",
            "Estado_VIF",
            "Estado_Final"
        ]
    ].head(20)

)
```

```python
#==========================================================================================
# JUSTIFICACIÓN METODOLÓGICA
#==========================================================================================

def justificar_estado(estado):

    justificaciones = {

        "Seleccionada":
        "Cumple todos los criterios del Framework.",

        "Pendiente por Cobertura":
        "Cobertura inferior al umbral mínimo establecido.",

        "Revisar por Multicolinealidad":
        "Presenta multicolinealidad perfecta.",

        "No Prioritaria":
        "IPML inferior al umbral definido.",

        "Excluida por Varianza Nula":
        "Variable constante que no aporta información al modelo."

    }

    return justificaciones.get(estado, "")


ranking["Justificacion"] = ranking["Estado_Final"].apply(justificar_estado)
```
