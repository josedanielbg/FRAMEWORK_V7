Este procedimiento permitió asociar correctamente cada registro climático con el valor mensual del índice ONI y su correspondiente clasificación ENSO, independientemente del día específico utilizado por cada fuente de datos.

```python
# ============================================================
# CREAR AÑO Y MES
# ============================================================

calidad["Anio"] = calidad["Fecha"].dt.year
calidad["Mes"] = calidad["Fecha"].dt.month
```

```python
# ============================================================
# TABLA MENSUAL DE CALIDAD DEL AGUA
# ============================================================

calidad_mensual = (

    calidad

    .pivot_table(

        index=["Nodo","Anio","Mes"],

        columns="propiedad_observada",

        values="resultado",

        aggfunc="mean"

    )

    .reset_index()

)

print(calidad_mensual.shape)

display(calidad_mensual.head())
```

```python
# ============================================================
# CONSERVAR SOLO LOS MUNICIPIOS DEL FRAMEWORK
# ============================================================

equivalencias = {

    "BOGOTÁ D.C.": "Bogota_Sabana",
    "CHÍA": "Bogota_Sabana",
    "TOCANCIPÁ": "Tocancipa_Media",
    "VILLAPINZÓN": "Villapinzon_Alta"

}

calidad_mensual = calidad_mensual[
    calidad_mensual["Nodo"].isin(equivalencias.keys())
].copy()

calidad_mensual["Nodo"] = (
    calidad_mensual["Nodo"]
    .replace(equivalencias)
)

print(calidad_mensual["Nodo"].unique())
print(calidad_mensual.shape)
```

```python
variables_calidad = [
    "CONDUCTIVIDAD ELECTRICA",
    "DEMANDA BIOQUIMICA DE OXIGENO (DBO5)",
    "DEMANDA QUIMICA DE OXIGENO (DQO)",
    "FOSFORO TOTAL",
    "NITROGENO TOTAL",
    "OXIGENO DISUELTO (OD)",
    "SOLIDOS SUSPENDIDOS TOTALES",
    "TEMPERATURA",
    "TURBIDEZ",
    "pH"
]

calidad_mensual = (
    calidad_mensual
    .groupby(
        ["Nodo","Anio","Mes"],
        as_index=False
    )[variables_calidad]
    .mean()
)
```

```python
calidad_mensual["Fecha"] = pd.to_datetime(
    dict(
        year=calidad_mensual["Anio"],
        month=calidad_mensual["Mes"],
        day=1
    )
)
```

```python
# ============================================================
# PASO 7 - INTEGRACIÓN CAPA CALIDAD DEL AGUA
# ============================================================

dataset_maestro = dataset_maestro.drop(
    columns=[
        "CONDUCTIVIDAD ELECTRICA",
        "DEMANDA BIOQUIMICA DE OXIGENO (DBO5)",
        "DEMANDA QUIMICA DE OXIGENO (DQO)",
        "FOSFORO TOTAL",
        "NITROGENO TOTAL",
        "OXIGENO DISUELTO (OD)",
        "SOLIDOS SUSPENDIDOS TOTALES",
        "TEMPERATURA",
        "TURBIDEZ",
        "pH"
    ],
    errors="ignore"
)

dataset_maestro = dataset_maestro.merge(
    calidad_mensual,
    on=["Nodo","Anio","Mes"],
    how="left"
)

print("\nIntegración realizada correctamente.")

print("\nDimensiones:")

print(dataset_maestro.shape)

display(dataset_maestro.head())
```

**Análisis Técnico**

**1. Objetivo**

Incorporar al Dataset Maestro la información de calidad del agua proveniente de la Red Nacional de Monitoreo Ambiental, con el propósito de complementar las variables climáticas, oceánicas, hidráulicas, hidrológicas, de gobernanza y percepción mediante indicadores fisicoquímicos representativos del estado de los cuerpos de agua.

Esta integración permite disponer de un conjunto de variables ambientales que describen simultáneamente las condiciones meteorológicas, hidráulicas, hidrológicas y de calidad del recurso hídrico, fortaleciendo la capacidad analítica del Framework para el modelado y la predicción mediante técnicas de Ciencia de Datos.

**2. Descripción del proceso**

La capa de Calidad del Agua se construyó a partir de registros históricos de monitoreo ambiental.
* **Filtro Espacial:** Inicialmente, la fuente contenía **10,875 registros** correspondientes a 68 propiedades de calidad del agua (cobertura temporal 2005–2024). Dado que el Framework se limita a la cuenca de estudio, se excluyeron los municipios que no pertenecen al ámbito geográfico del proyecto (`Cota`, `Soacha`, `Girardot`, `El Colegio` y `Tocaima`), conservando únicamente los registros de `Bogotá D.C.`, `Chía`, `Tocancipá` y `Villapinzón`.
* **Homologación de Nodos:** Se mapearon los municipios seleccionados hacia la nomenclatura oficial del Dataset Maestro:

| Municipio original | Nodo Framework (Destino) |
| :--- | :--- |
| `Bogotá D.C.` | **Bogota_Sabana** |
| `Chía` | **Bogota_Sabana** |
| `Tocancipá` | **Tocancipa_Media** |
| `Villapinzón` | **Villapinzon_Alta** |

**3. Selección de variables**

De los 68 parámetros fisicoquímicos iniciales, se seleccionaron los **10 atributos de mayor relevancia técnica** y densidad temporal de muestreo:

* Conductividad eléctrica
* Demanda Química de Oxígeno (DQO)
* Demanda Bioquímica de Oxígeno ($DBO_5$)
* Oxígeno Disuelto (OD)
* Temperatura del agua
* pH
* Turbidez
* Fósforo Total
* Nitrógeno Total
* Sólidos Suspendidos Totales (SST)

**4. Transformación temporal**

La estructura transaccional original (formato longitudinal) requirió un preprocesamiento sistemático antes de la fusión:
* **Agregación:** Se agruparon las observaciones por Nodo, Año, Mes y propiedad, calculando el valor promedio mensual.
* **Pivotado (Long-to-Wide):** Se reestructuraron los datos para pasar de un formato de filas transaccionales a un formato matricial amplio, generando una columna única para cada parámetro.

**5. Resultado del preprocesamiento:**

Se consolidó una matriz temporalmente alineada de **137 registros y 13 columnas**, optimizada para la integración de datos.

**6. Integración con el Dataset Maestro**
La fusión se ejecutó aplicando un **`LEFT JOIN`** utilizando como claves combinadas de unión:

$$\text{Llaves de unión: } \{\text{Nodo, Año, Mes}\}$$

Esta operación prioriza la permanencia de todos los registros preexistentes del Dataset Maestro, anexando las propiedades físico-químicas de calidad del agua únicamente en los periodos y nodos donde se ejecutaron campañas de monitoreo efectivas.

**7. Resultados de la validación**

La auditoría de calidad posterior al proceso de integración confirmó los siguientes puntos de control:

* **Consistencia estructural:** El número total de registros del Dataset Maestro permaneció inalterado (528 registros), confirmando que el proceso no indujo duplicaciones accidentales de filas.
* **Alineación espacial y temporal:** Las mediciones del laboratorio fueron asignadas con precisión matemática a sus correspondientes nodos y meses.

**8. Justificación de Datos Faltantes (NaN)**

La presencia de valores nulos observada es representativa de la dinámica de captura de este dominio científico y responde a los siguientes factores de campo:
* **Muestreo Discontinuo:** A diferencia de las variables climáticas de registro continuo diario, las variables de calidad del agua se miden de forma periódica en campañas de muestreo específicas (no mensuales).
* **Ausencia de Estaciones en Sopo_Sur:** El nodo **Sopo_Sur** carece de datos debido a que la red oficial no dispone de estaciones físicas de monitoreo activas en ese tramo de control durante el periodo evaluado.
* **Límite Temporal (Año 2025):** Las series históricas de la fuente oficial de calidad del agua finalizan en el año 2024, razón por la cual los registros correspondientes al 2025 permanecen nulos.
