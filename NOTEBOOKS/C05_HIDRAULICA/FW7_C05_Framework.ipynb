# **CAPA HIDRAULICA EN MASA**


# **Objetivo de la Capa**

Consolidar y estandarizar la información hidráulica correspondiente al volumen útil almacenado en los embalses de la cuenca del río Bogotá, mediante la transformación de registros diarios en una serie temporal mensual, garantizando su calidad, trazabilidad e interoperabilidad para su integración con las demás capas del Framework V7 y su utilización como variable explicativa en modelos predictivos de calidad del agua basados en redes neuronales LSTM.

#  **Justificación Científica**

La dinámica hidráulica de una cuenca constituye uno de los principales factores que condicionan la disponibilidad, permanencia y renovación del recurso hídrico. Las variaciones en el volumen almacenado de los embalses modifican el régimen de caudales, los tiempos de residencia del agua y los procesos de mezcla, influyendo directamente sobre parámetros fisicoquímicos como el oxígeno disuelto, la conductividad eléctrica, la turbidez y la concentración de nutrientes.

Desde la perspectiva del aprendizaje profundo, el volumen útil almacenado representa una variable de estado del sistema hidrológico, capaz de describir la disponibilidad real del recurso en un instante determinado y su evolución temporal. Su incorporación permite que los modelos LSTM capturen relaciones de dependencia entre el almacenamiento hidráulico y las variaciones observadas en la calidad del agua, fortaleciendo la capacidad predictiva frente a cambios hidrológicos de mediano y largo plazo.

Para garantizar la coherencia temporal con las demás fuentes de información del Framework, los registros diarios publicados por XM fueron consolidados mediante promedios mensuales del indicador VolumenUtilDiarioMasa, seleccionando exclusivamente el embalse AGREGADO_BOGOTA, por representar el comportamiento integrado del sistema de almacenamiento que abastece la cuenca objeto de estudio. Esta transformación reduce la variabilidad diaria sin perder la tendencia hidrológica, facilitando la integración con las variables climáticas, meteorológicas y de calidad del agua.

# **Alcance de la capa**

La capa hidráulica comprende el proceso completo de adquisición, transformación y estandarización de la información oficial de reservas hidráulicas publicada por XM, incluyendo:

Incorporación de metadatos, identificadores únicos, auditoría, indicadores de calidad y mecanismos de trazabilidad mediante hash.
Generación del conjunto de datos estandarizado 02_Capa_Hidraulica_Framework.xlsx, listo para su integración con las capas de calidad del agua, información climática, variables meteorológicas y demás componentes del Dataset Maestro utilizado en el entrenamiento del modelo LSTM.

En consecuencia, esta capa constituye el componente hidráulico del Framework V7, proporcionando una representación temporal consistente del estado de almacenamiento del sistema de embalses que permite explicar parte de la variabilidad hidrológica que afecta la calidad del agua en la cuenca del río Bogotá y fortalece la construcción de modelos predictivos orientados a la gestión sostenible del recurso hídrico.

## **M0 – Configuración**

````python
import pandas as pd
import numpy as np
import hashlib
from datetime import datetime
import os

# ============================================================
# FRAMEWORK V7
# FRAMEWORK CORE
# GENERADOR UNIVERSAL DE CAPAS
# ============================================================

NOMBRE_CAPA = "Hidráulica"

CODIGO_CAPA = "HDR"

VERSION = "1.0"

RESPONSABLE = "Juan Carlos Riatiga"

FUENTE = "Reservas Hidráulicas en Masa - XM"

URL_FUENTE = "https://www.datos.gov.co/dataset/Reservas-Hidr-ulicas-en-Masa/5cbd-e6b5"

ARCHIVO = "https://raw.githubusercontent.com/jriatiga/dataset/refs/heads/main/proyectoGradoV4/05_Capa_Hidraulica/01_Capa_Reservas_Hidraulicas_Masa_V1.csv"
````

## **M1 – Lectura**

````python
import pandas as pd

print("="*70)
print("FRAMEWORK V7 - CAPA HIDRÁULICA")
print("M1 - LECTURA DEL DATASET")
print("="*70)

# ============================================================
# Lectura del archivo desde GitHub
# ============================================================

df = pd.read_csv(ARCHIVO)

print("\nArchivo cargado correctamente.\n")

# ============================================================
# Dimensiones
# ============================================================

print("="*70)
print("DIMENSIONES")
print("="*70)
print(df.shape)

# ============================================================
# Columnas
# ============================================================

print("\n" + "="*70)
print("COLUMNAS")
print("="*70)

for columna in df.columns:
    print(columna)

# ============================================================
# Primeros registros
# ============================================================

print("\n" + "="*70)
print("PRIMEROS REGISTROS")
print("="*70)

print(df.head())

# ============================================================
# Últimos registros
# ============================================================

print("\n" + "="*70)
print("ÚLTIMOS REGISTROS")
print("="*70)

print(df.tail())
````

## **M2 – Validación estructural**

````python
print("="*70)
print("M2 - VALIDACIÓN ESTRUCTURAL")
print("="*70)

# ============================================================
# Tipos de datos originales
# ============================================================

print("\nTipos de datos originales:\n")
print(df.dtypes)

# ============================================================
# Validación de la columna Fecha
# ============================================================

print("\nTipo actual de 'Fecha':")
print(type(df.loc[0, "Fecha"]))

print("\nPrimeros valores de Fecha:")
print(df["Fecha"].head())

# ============================================================
# Conversión de fechas
# ============================================================

df["Fecha"] = pd.to_datetime(df["Fecha"])
df["FechaPublicacion"] = pd.to_datetime(df["FechaPublicacion"])

# ============================================================
# Verificación posterior
# ============================================================

print("\nDespués de la conversión:\n")
print(df[["FechaPublicacion","Fecha"]].dtypes)

print("\nPrimeros registros convertidos:")
print(df[["FechaPublicacion","Fecha"]].head())
````

## **M3 – Estandarización**

````python
print("="*70)
print("M3 - ESTANDARIZACIÓN")
print("="*70)

# ============================================================
# 1. Filtrar únicamente el sistema Bogotá
# ============================================================

df = df[df["CodigoEmbalse"] == "AGREGADO_BOGOTA"].copy()

print("\nRegistros después del filtro:")
print(df.shape)

# ============================================================
# 2. Conservar únicamente las variables necesarias
# ============================================================

df = df[
    [
        "Fecha",
        "CodigoEmbalse",
        "VolumenUtilDiarioMasa"
    ]
].copy()

print("\nColumnas conservadas:")
print(df.columns.tolist())

# ============================================================
# 3. Crear identificador mensual
# ============================================================

df["mes_id"] = df["Fecha"].dt.to_period("M")

# ============================================================
# 4. Consolidación mensual
# ============================================================

df = (
    df.groupby(
        ["mes_id", "CodigoEmbalse"],
        as_index=False
    )["VolumenUtilDiarioMasa"]
    .mean()
)

# ============================================================
# 5. Convertir nuevamente la fecha
# ============================================================

df["Fecha"] = df["mes_id"].dt.to_timestamp("M")

df.drop(columns="mes_id", inplace=True)

# ============================================================
# 6. Ordenar cronológicamente
# ============================================================

df = df.sort_values("Fecha").reset_index(drop=True)

# ============================================================
# Resultado
# ============================================================

print("\nDimensiones finales:")
print(df.shape)

print("\nPrimeros registros:")
print(df.head())

print("\nÚltimos registros:")
print(df.tail())
````

¿Por qué utilizar el promedio mensual?

En este caso, el promedio mensual de VolumenUtilDiarioMasa es la opción más adecuada porque:

* Reduce la variabilidad diaria sin perder la tendencia hidrológica.

* Mantiene la coherencia temporal con las demás capas (ONI, hidrología y calidad del agua), que también se analizan a escala mensual.

* Proporciona una serie estable para el entrenamiento del modelo LSTM, evitando que fluctuaciones diarias introduzcan ruido innecesario.

Con este M3, la Capa Hidráulica queda estandarizada y lista para continuar con el M4 – Gobierno del Dato, donde incorporaremos los identificadores del Framework (ID_Registro, ID_Capa), la trazabilidad y el hash, siguiendo el mismo estándar aplicado en las capas anteriores.

## **M4 – Gobierno del Dato**

````python
print("="*70)
print("M4 - GOBIERNO DEL DATO")
print("="*70)

# ============================================================
# 1. ID único del registro
# ============================================================

df.insert(
    0,
    "ID_Registro",
    [
        f"{CODIGO_CAPA}-{str(i+1).zfill(7)}"
        for i in range(len(df))
    ]
)

# ============================================================
# 2. Código de la capa
# ============================================================

df.insert(1, "ID_Capa", CODIGO_CAPA)

# ============================================================
# 3. Código estación (embalse)
# ============================================================

df["codigoestacion"] = df["CodigoEmbalse"]

# ============================================================
# 4. Fecha ETL
# ============================================================

df["Fecha_ETL"] = datetime.now().date()

# ============================================================
# 5. Hora ETL
# ============================================================

df["Hora_ETL"] = datetime.now().strftime("%H:%M:%S")

# ============================================================
# 6. Versión Framework
# ============================================================

df["Version_Framework"] = VERSION

# ============================================================
# 7. Estado del registro
# ============================================================

df["Estado_Registro"] = "VALIDADO"

# ============================================================
# 8. Responsable
# ============================================================

df["Responsable"] = RESPONSABLE

# ============================================================
# 9. Hash del registro
# ============================================================

def generar_hash(fila):

    texto = "|".join(map(str, fila.values))

    return hashlib.md5(texto.encode()).hexdigest()

df["Hash"] = df.apply(generar_hash, axis=1)

# ============================================================
# Resultado
# ============================================================

print("\nDimensiones finales:")
print(df.shape)

print("\nPrimeros registros:")
print(df.head())

print("\nColumnas finales:")
print(df.columns.tolist())
````
