# **CAPA HIDROLOGICA**

# **Objetivo de la Capa Hidrológica**

La Capa Hidrológica tiene como finalidad incorporar información observada sobre los niveles del agua registrados por estaciones hidrométricas ubicadas dentro de la cuenca del río Bogotá, con el propósito de representar el comportamiento físico del sistema hídrico y proporcionar una base objetiva para el análisis temporal, la interoperabilidad con otras capas del Framework y el desarrollo de modelos de apoyo a la toma de decisiones.

Esta capa constituye la representación directa del estado del recurso hídrico, permitiendo relacionar la dinámica del río con las condiciones climáticas, la calidad del agua, los eventos extremos, las reservas hidráulicas, la gobernanza y los indicadores institucionales.


#  **Justificación Científica**

A diferencia de la Capa Climática, que representa los factores atmosféricos que influyen sobre el recurso hídrico, la Capa Hidrológica describe la respuesta física del sistema.

Los niveles del río constituyen una de las variables más importantes para:

* evaluar disponibilidad hídrica;
* identificar eventos extremos;
* analizar periodos de creciente y estiaje;
* estudiar la respuesta del río frente a variaciones climáticas;
* alimentar modelos predictivos de inundaciones;
* apoyar procesos de gestión integral del recurso hídrico.

Dentro del Framework esta capa actuará como una de las variables centrales para la integración multifuente.

# **Alcance de la capa**

La Capa Hidrológica integra variables relacionadas con el comportamiento y la disponibilidad del recurso hídrico superficial para los municipios de Chía, Sopó, Tocancipá y Villapinzón, consolidando información histórica proveniente de fuentes oficiales previamente depuradas y estandarizadas.

Su alcance comprende la organización de registros temporales asociados a cada nodo territorial, garantizando consistencia en la estructura, calidad, trazabilidad y gobierno del dato mediante procesos de validación, normalización, generación de metadatos, auditoría e indicadores de calidad.

Esta capa proporciona una representación homogénea del estado hidrológico de los municipios, permitiendo:

caracterizar la dinámica del recurso hídrico mediante variables hidrológicas relevantes;
servir como insumo para la integración con las capas Climática, Percepción y Gobernanza dentro del Framework de interoperabilidad;
facilitar análisis espacio-temporales sobre la disponibilidad y comportamiento del agua;
proporcionar variables de entrada para modelos analíticos, predictivos y de aprendizaje automático, incluyendo redes neuronales LSTM y otros algoritmos de Ciencia de Datos orientados al apoyo de la gestión hídrica.

La cobertura temporal de la capa corresponde al período disponible en las fuentes oficiales utilizadas, preservando la resolución temporal original y garantizando la interoperabilidad con las demás capas del Framework mediante una estructura unificada basada en nodo y fecha.

## **M3 – Estandarización**

````python
#=========================================================
# CREAR IDENTIFICADORES
#=========================================================

# Check if 'ID_Registro' column already exists before inserting
if "ID_Registro" not in df.columns:
    df.insert(
        0,
        "ID_Registro",
        [
            f"{CODIGO_CAPA}-{str(i).zfill(7)}"
            for i in range(1, len(df)+1)
        ]
    )
else:
    print("Column 'ID_Registro' already exists, skipping insertion.")

# Check if 'ID_Capa' column already exists before creating
if "ID_Capa" not in df.columns:
    df["ID_Capa"] = CODIGO_CAPA
else:
    print("Column 'ID_Capa' already exists, skipping creation.")

print("Identificadores creados.")
````

````python
#=========================================================
# ORDENAR COLUMNAS
#=========================================================

columnas_control = [

    "ID_Registro",

    "ID_Capa"

]

otras_columnas = [

    c for c in df.columns

    if c not in columnas_control

]

df = df[columnas_control + otras_columnas]

print(df.columns)
````

````python
#=========================================================
# NORMALIZACIÓN DE TIPOS
#=========================================================

df["nombreestacion"] = df["nombreestacion"].astype(str)

df["departamento"] = df["departamento"].astype(str)

df["municipio"] = df["municipio"].astype(str)

df["zonahidrografica"] = df["zonahidrografica"].astype(str)

df["descripcionsensor"] = df["descripcionsensor"].astype(str)

df["unidadmedida"] = df["unidadmedida"].astype(str)

print("Tipos normalizados.")
````

````python
#=========================================================
# LIMPIEZA DE TEXTO
#=========================================================

df.columns = df.columns.str.strip()

for c in [

    "nombreestacion",

    "departamento",

    "municipio",

    "zonahidrografica",

    "descripcionsensor",

    "unidadmedida"

]:

    df[c] = df[c].str.strip()

print("Espacios eliminados.")
````

````python
#=========================================================
# ORDENAR REGISTROS
#=========================================================

df = df.sort_values(

    by=[

        "nombreestacion",

        "fechaobservacion"

    ]

).reset_index(drop=True)

print("Registros ordenados.")
````

````python
#=========================================================
# VERIFICACIÓN
#=========================================================

print(df.head())

print()

print(df.columns)

print()

print(df.dtypes)
````

## **M4 – Gobierno del Dato**

````python
#=========================================================
# FECHA ETL
#=========================================================

fecha = datetime.now()

df["Fecha_ETL"] = fecha.strftime("%Y-%m-%d")

df["Hora_ETL"] = fecha.strftime("%H:%M:%S")

print("Fecha ETL registrada.")
````

````python
#=========================================================
# VERSIÓN
#=========================================================

df["Version_Framework"] = VERSION

print("Versión asignada.")
````

````python
#=========================================================
# RESPONSABLE
#=========================================================

df["Responsable"] = RESPONSABLE

print("Responsable asignado.")
````

````python
#=========================================================
# HASH DEL REGISTRO
#=========================================================

def generar_hash(fila):

    texto = "".join(fila.astype(str))

    return hashlib.md5(texto.encode()).hexdigest()

df["Hash"] = df.astype(str).apply(generar_hash, axis=1)

print("Hash generado.")
````

````python
#=========================================================
# ESTADO DEL REGISTRO
#=========================================================

df["Estado_Registro"] = "VALIDADO"

print("Estado asignado.")
````

````python
#=========================================================
# VERIFICACIÓN
#=========================================================

print(df.head())

print()

print(df.columns)

print()

print(df.dtypes)
````

## **M5 – Metadata**

````python
#=========================================================
# FRAMEWORK V7
# M5 - METADATA
#=========================================================

metadata = pd.DataFrame({

    "Propiedad":[

        "Nombre Capa",
        "Código",
        "Fuente",
        "URL",
        "Versión",
        "Responsable",
        "Fecha ETL",
        "Registros",
        "Variables"

    ],

    "Valor":[

        NOMBRE_CAPA,
        CODIGO_CAPA,
        FUENTE,
        URL_FUENTE,
        VERSION,
        RESPONSABLE,
        fecha.strftime("%Y-%m-%d"),
        len(df),
        len(df.columns)

    ]

})

metadata.to_excel(
    "03_Metadata.xlsx",
    index=False
)

print(metadata)

print()

print("Metadata generada correctamente.")
````

## **M6 – Diccionario de Datos**

````python
#=========================================================
# FRAMEWORK V7
# M6 - DICCIONARIO DE DATOS
#=========================================================

diccionario=[]

for columna in df.columns:

    diccionario.append({

        "Campo":columna,

        "Tipo":str(df[columna].dtype),

        "Nulos":df[columna].isnull().sum(),

        "Valores_Unicos":df[columna].nunique(),

        "Descripción":""

    })

diccionario=pd.DataFrame(diccionario)

diccionario.to_excel(

    "04_Diccionario_Datos.xlsx",

    index=False

)

print(diccionario)

print()

print("Diccionario generado correctamente.")
````

## **M7 – Auditoría**

````python
#=========================================================
# FRAMEWORK V7
# M7 - AUDITORÍA
#=========================================================

auditoria = pd.DataFrame({

    "Indicador":[

        "Registros",
        "Variables",
        "Valores Nulos",
        "Duplicados",
        "Fecha Auditoría"

    ],

    "Valor":[

        len(df),

        len(df.columns),

        df.isnull().sum().sum(),

        df.duplicated().sum(),

        fecha.strftime("%Y-%m-%d")

    ]

})

auditoria.to_excel(

    "05_Auditoria.xlsx",

    index=False

)

print(auditoria)

print()

print("Auditoría generada correctamente.")
````

## **M8 – Indicadores**

````python
#=========================================================
# FRAMEWORK V7
# M8 - INDICADORES
#=========================================================

import numpy as np

total_registros = len(df)

total_variables = len(df.columns)

total_nulos = df.isnull().sum().sum()

duplicados = df.duplicated().sum()

variables_numericas = len(

    df.select_dtypes(include=np.number).columns

)

integridad = round(

    (

        (total_registros*total_variables)-total_nulos

    )/

    (total_registros*total_variables)

    *100

,2)

porcentaje_nulos = round(

    total_nulos/

    (total_registros*total_variables)

    *100

,2)

#-----------------------------------------

fecha_inicio = df["fechaobservacion"].min()

fecha_fin = df["fechaobservacion"].max()

#-----------------------------------------

numero_estaciones = df["nombreestacion"].nunique()

numero_municipios = df["municipio"].nunique()

numero_departamentos = df["departamento"].nunique()

#-----------------------------------------

estado = "APROBADO"

if total_nulos>0:

    estado="REVISAR"

if duplicados>0:

    estado="REVISAR"

#-----------------------------------------

indicadores = pd.DataFrame({

"Indicador":[

"Nombre Capa",

"Código",

"Versión",

"Fuente",

"Fecha",

"Registros",

"Variables",

"Variables Numéricas",

"Nulos",

"% Nulos",

"Duplicados",

"Integridad",

"Inicio",

"Fin",

"Estaciones",

"Municipios",

"Departamentos",

"Estado"

],

"Valor":[

NOMBRE_CAPA,

CODIGO_CAPA,

VERSION,

FUENTE,

fecha.strftime("%Y-%m-%d"),

total_registros,

total_variables,

variables_numericas,

total_nulos,

porcentaje_nulos,

duplicados,

integridad,

fecha_inicio,

fecha_fin,

numero_estaciones,

numero_municipios,

numero_departamentos,

estado

]

})

indicadores.to_excel(

"06_Indicadores.xlsx",

index=False

)

print(indicadores)
````
