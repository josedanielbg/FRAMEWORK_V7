# **CAPA PERCEPCION**


# **Objetivo de la Capa**

La Capa de Percepción tiene como objetivo consolidar y estandarizar la información relacionada con la percepción de la calidad del agua para los municipios de Bogotá D.C., Chía, Tocancipá y Villapinzón, utilizando el Índice de Riesgo de la Calidad del Agua para Consumo Humano (IRCA) como indicador representativo.

Esta capa proporciona una visión temporal del comportamiento de la calidad del agua percibida desde el punto de vista sanitario, permitiendo su integración con las demás capas del Framework para apoyar el análisis multidimensional de la gestión hídrica mediante técnicas de Ciencia de Datos e Inteligencia Artificial.

#  **Justificación Científica**

La gestión sostenible del recurso hídrico requiere incorporar no solamente variables ambientales e hidrológicas, sino también indicadores que reflejen el impacto del agua sobre la población y las condiciones de consumo humano.

El Índice de Riesgo de la Calidad del Agua para Consumo Humano (IRCA), establecido por la normativa colombiana, constituye uno de los principales indicadores utilizados para evaluar la calidad sanitaria del agua suministrada a la población. Su incorporación dentro del Framework permite representar una dimensión complementaria del sistema hídrico, integrando aspectos relacionados con la salud pública y el nivel de riesgo asociado al abastecimiento de agua potable.

Desde la perspectiva de Ciencia de Datos, esta información enriquece el conjunto de variables disponibles para el análisis exploratorio, la construcción de modelos predictivos y el estudio de relaciones entre las condiciones climáticas, hidrológicas, institucionales y la calidad del agua, fortaleciendo los procesos de interoperabilidad entre las diferentes capas del Framework.

# **Alcance de la capa**

La Capa de Percepción integra información histórica del Índice de Riesgo de la Calidad del Agua para Consumo Humano (IRCA) correspondiente a los municipios de Bogotá D.C., Chía, Tocancipá y Villapinzón, organizando los registros por nodo y período de análisis.

Su alcance comprende los procesos de adquisición, validación, depuración, estandarización y gobierno del dato, garantizando la consistencia estructural, la trazabilidad y la interoperabilidad con las demás capas del Framework.

La información consolidada permite:

* caracterizar la evolución temporal de la calidad del agua para consumo humano;
* identificar cambios en los niveles de riesgo sanitario durante el período de estudio;
* integrar indicadores de percepción con variables climáticas, hidrológicas y de gobernanza;
* servir como insumo para modelos analíticos y predictivos orientados al apoyo de la toma de decisiones en la gestión sostenible del recurso hídrico.

La capa conserva la resolución temporal de la fuente oficial utilizada y adopta la estructura homogénea definida por el Framework, facilitando su integración con el Dataset Maestro y los posteriores procesos de análisis mediante algoritmos de Ciencia de Datos e Inteligencia Artificial.

## **M3 – Estandarización**

````python
# ============================================================
# M3 - ESTANDARIZACIÓN
# FRAMEWORK V7
# ============================================================

print("="*70)
print("M3 - ESTANDARIZACIÓN")
print("="*70)

# ============================================================
# Copia del dataset
# ============================================================

df_std = df.copy()

# ============================================================
# Normalización de nombres
# ============================================================

df_std.columns = [
    c.strip().replace(" ", "_")
    for c in df_std.columns
]

# ============================================================
# Conversión de tipos
# ============================================================

df_std["Nodo"] = df_std["Nodo"].astype(str)

df_std["municipio"] = df_std["municipio"].astype(str)

df_std["a_o"] = df_std["a_o"].astype(int)

df_std["irca"] = pd.to_numeric(
    df_std["irca"],
    errors="coerce"
)

df_std["nivel_de_riesgo"] = (
    df_std["nivel_de_riesgo"]
    .astype(str)
    .str.strip()
)

# ============================================================
# Crear Fecha (31 de diciembre de cada año)
# ============================================================

df_std["Fecha"] = pd.to_datetime(
    df_std["a_o"].astype(str) + "-12-31"
)

# ============================================================
# Ordenamiento
# ============================================================

df_std = df_std.sort_values(
    ["Nodo","Fecha"]
).reset_index(drop=True)

# ============================================================
# Auditoría
# ============================================================

print()

print("Primeros registros")
print(df_std.head())

print()

print("Tipos de datos")
print(df_std.dtypes)

print()

print("Valores nulos")
print(df_std.isnull().sum())

print()

print("Cobertura temporal")

print(
    df_std["Fecha"].min(),
    "->",
    df_std["Fecha"].max()
)

print()

print("Nodos")

print(df_std["Nodo"].value_counts())

print()

print("Niveles de riesgo")

print(df_std["nivel_de_riesgo"].value_counts())

print()

print("Dataset estandarizado correctamente.")
````

## **M4 – Gobierno del Dato**

````python
# ============================================================
# M4 - GOBIERNO DEL DATO
# FRAMEWORK V7
# ============================================================

print("="*70)
print("M4 - GOBIERNO DEL DATO")
print("="*70)

# ============================================================
# Copia del dataset
# ============================================================

df_gob = df_std.copy()

# ============================================================
# Identificador único
# ============================================================

df_gob.insert(
    0,
    "ID_Registro",
    [
        f"{CODIGO_CAPA}-{str(i+1).zfill(7)}"
        for i in range(len(df_gob))
    ]
)

# ============================================================
# Identificador de capa
# ============================================================

df_gob.insert(
    1,
    "ID_Capa",
    CODIGO_CAPA
)

# ============================================================
# Código de estación (Nodo)
# ============================================================

df_gob["codigoestacion"] = df_gob["Nodo"]

# ============================================================
# Metadatos ETL
# ============================================================

fecha_actual = datetime.now()

df_gob["Fecha_ETL"] = fecha_actual.strftime("%Y-%m-%d")

df_gob["Hora_ETL"] = fecha_actual.strftime("%H:%M:%S")

df_gob["Version_Framework"] = VERSION

df_gob["Estado_Registro"] = "VALIDADO"

df_gob["Responsable"] = RESPONSABLE

# ============================================================
# Hash de integridad
# ============================================================

def generar_hash(fila):

    texto = "|".join(
        map(str, fila.values)
    )

    return hashlib.md5(
        texto.encode("utf-8")
    ).hexdigest()

df_gob["Hash"] = df_gob.apply(
    generar_hash,
    axis=1
)

# ============================================================
# Auditoría
# ============================================================

print()

print("Primeros registros")

print(df_gob.head())

print()

print("Columnas")

print(df_gob.columns.tolist())

print()

print("Valores nulos")

print(df_gob.isnull().sum())

print()

print("Hashes únicos")

print(df_gob["Hash"].nunique())

print()

print("Total registros")

print(len(df_gob))

print()

print("="*70)
print("GOBIERNO DEL DATO COMPLETADO")
print("="*70)
````

## **M5 – Metadata**

````python
# ============================================================
# M5 - METADATA
# FRAMEWORK V7
# ============================================================

print("="*70)
print("M5 - METADATA")
print("="*70)

# ============================================================
# Construcción del Metadata
# ============================================================

metadata = []

for columna in df_gob.columns:

    metadata.append({

        "Campo": columna,

        "Tipo_Dato": str(df_gob[columna].dtype),

        "Nulos": int(df_gob[columna].isnull().sum()),

        "No_Nulos": int(df_gob[columna].notnull().sum()),

        "Duplicados": int(df_gob[columna].duplicated().sum()),

        "Valores_Unicos": int(df_gob[columna].nunique()),

        "Ejemplo": str(df_gob[columna].dropna().iloc[0]) if df_gob[columna].notnull().sum() > 0 else ""

    })

df_metadata = pd.DataFrame(metadata)

# ============================================================
# Exportación
# ============================================================

archivo_metadata = "03_Metadata.xlsx"

df_metadata.to_excel(
    archivo_metadata,
    index=False
)

print()

print("Archivo generado correctamente:")

print(archivo_metadata)

print()

print(df_metadata)
````

## **M8 – Indicadores**

````python
# ============================================================
# M8 - INDICADORES
# FRAMEWORK V7
# ============================================================

print("="*70)
print("M8 - INDICADORES")
print("="*70)

# ============================================================
# Indicadores generales
# ============================================================

total_registros = len(df_gob)

total_variables = len(df_gob.columns)

total_nodos = df_gob["Nodo"].nunique()

fecha_inicio = df_gob["Fecha"].min()

fecha_fin = df_gob["Fecha"].max()

irca_promedio = df_gob["irca"].mean()

irca_maximo = df_gob["irca"].max()

irca_minimo = df_gob["irca"].min()

riesgo_dominante = df_gob["nivel_de_riesgo"].mode()[0]

sin_riesgo = (df_gob["nivel_de_riesgo"]=="Sin riesgo").sum()

riesgo_bajo = (df_gob["nivel_de_riesgo"]=="Riesgo bajo").sum()

riesgo_medio = (df_gob["nivel_de_riesgo"]=="Riesgo medio").sum()

riesgo_alto = (df_gob["nivel_de_riesgo"]=="Riesgo alto").sum()

riesgo_inviable = (df_gob["nivel_de_riesgo"]=="Inviable sanitariamente").sum()

porcentaje_sin_riesgo = round((sin_riesgo/total_registros)*100,2)

# ============================================================
# Construcción del DataFrame
# ============================================================

indicadores = pd.DataFrame({

    "Indicador":[

        "Total Registros",
        "Total Variables",
        "Total Nodos",
        "Fecha Inicial",
        "Fecha Final",
        "IRCA Promedio",
        "IRCA Máximo",
        "IRCA Mínimo",
        "Nivel de Riesgo Dominante",
        "Registros Sin Riesgo",
        "Registros Riesgo Bajo",
        "Registros Riesgo Medio",
        "Registros Riesgo Alto",
        "Registros Inviable Sanitariamente",
        "% Sin Riesgo"

    ],

    "Valor":[

        total_registros,
        total_variables,
        total_nodos,
        str(fecha_inicio.date()),
        str(fecha_fin.date()),
        round(irca_promedio,2),
        round(irca_maximo,2),
        round(irca_minimo,2),
        riesgo_dominante,
        sin_riesgo,
        riesgo_bajo,
        riesgo_medio,
        riesgo_alto,
        riesgo_inviable,
        porcentaje_sin_riesgo

    ]

})

# ============================================================
# Exportación
# ============================================================

archivo = "06_Indicadores.xlsx"

indicadores.to_excel(
    archivo,
    index=False
)

print()

print("Archivo generado correctamente:")

print(archivo)

print()

print(indicadores)
````

## **M9 – Análisis Exploratorio de Datos (EDA)**

````python
# ============================================================
# M9 - ANÁLISIS EXPLORATORIO DE DATOS (EDA)
# FRAMEWORK V7
# ============================================================

print("="*70)
print("M9 - ANÁLISIS EXPLORATORIO DE DATOS (EDA)")
print("="*70)

# ============================================================
# Estadísticas generales
# ============================================================

print("\nESTADÍSTICAS DEL IRCA")
print("-"*70)

print(df_gob["irca"].describe())

# ============================================================
# Distribución por nivel de riesgo
# ============================================================

print("\nDISTRIBUCIÓN DEL NIVEL DE RIESGO")
print("-"*70)

riesgo = (
    df_gob["nivel_de_riesgo"]
    .value_counts()
    .sort_index()
)

print(riesgo)

# ============================================================
# Distribución por nodo
# ============================================================

print("\nREGISTROS POR NODO")
print("-"*70)

print(
    df_gob["Nodo"]
    .value_counts()
)

# ============================================================
# Promedio IRCA por nodo
# ============================================================

print("\nPROMEDIO DEL IRCA POR NODO")
print("-"*70)

irca_nodo = (
    df_gob
    .groupby("Nodo")["irca"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print(irca_nodo)

# ============================================================
# Promedio IRCA por año
# ============================================================

print("\nPROMEDIO DEL IRCA POR AÑO")
print("-"*70)

irca_anual = (
    df_gob
    .groupby(df_gob["Fecha"].dt.year)["irca"]
    .mean()
    .round(2)
)

print(irca_anual)

# ============================================================
# Cruce Nodo vs Riesgo
# ============================================================

print("\nTABLA CRUZADA NODO vs NIVEL DE RIESGO")
print("-"*70)

tabla = pd.crosstab(
    df_gob["Nodo"],
    df_gob["nivel_de_riesgo"]
)

print(tabla)

# ============================================================
# Variabilidad del IRCA
# ============================================================

print("\nVARIABILIDAD DEL IRCA")
print("-"*70)

print(f"Media               : {df_gob['irca'].mean():.2f}")
print(f"Desv. estándar      : {df_gob['irca'].std():.2f}")
print(f"Coef. Variación (%) : {(df_gob['irca'].std()/df_gob['irca'].mean())*100:.2f}")

# ============================================================
# Valores extremos
# ============================================================

print("\nVALORES EXTREMOS")
print("-"*70)

print("IRCA mínimo")
print(
    df_gob.loc[
        df_gob["irca"].idxmin(),
        ["Nodo","Fecha","irca","nivel_de_riesgo"]
    ]
)

print()

print("IRCA máximo")
print(
    df_gob.loc[
        df_gob["irca"].idxmax(),
        ["Nodo","Fecha","irca","nivel_de_riesgo"]
    ]
)

# ============================================================
# Conclusiones automáticas
# ============================================================

print("\nCONCLUSIONES DEL EDA")
print("-"*70)

print(f"• Registros analizados: {len(df_gob)}")

print(f"• Nodos evaluados: {df_gob['Nodo'].nunique()}")

print(f"• Cobertura temporal: {df_gob['Fecha'].min().year} - {df_gob['Fecha'].max().year}")

print(f"• IRCA promedio: {df_gob['irca'].mean():.2f}")

print(f"• Riesgo predominante: {df_gob['nivel_de_riesgo'].mode()[0]}")

print(f"• Máximo IRCA observado: {df_gob['irca'].max():.2f}")

print(f"• Mínimo IRCA observado: {df_gob['irca'].min():.2f}")

print()

print("EDA finalizado correctamente.")
````
