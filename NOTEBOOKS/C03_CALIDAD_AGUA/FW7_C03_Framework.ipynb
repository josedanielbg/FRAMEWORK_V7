# **CAPA CALIDAD DE AGUA**



# **Objetivo de la Capa**

Construir una capa de información de calidad del agua mediante la integración, estandarización, validación y gobierno de datos provenientes de la red oficial de monitoreo de la Corporación Autónoma Regional (CAR), con el propósito de consolidar un conjunto de datos confiable que permita caracterizar el estado fisicoquímico del río Bogotá y suministrar información para procesos de análisis estadístico, modelado predictivo e inteligencia artificial dentro del Framework de Gestión Hídrica.

La capa integra información histórica de parámetros fisicoquímicos y ambientales registrados en diferentes estaciones de monitoreo distribuidas a lo largo de la cuenca del río Bogotá, preservando la trazabilidad y la calidad de los datos durante todo el proceso de transformación.

#  **Justificación Científica**

La calidad del agua constituye uno de los componentes fundamentales para evaluar el estado ambiental de una cuenca hidrográfica, ya que refleja la interacción entre los procesos naturales y las actividades antrópicas desarrolladas sobre el territorio. Variables como el oxígeno disuelto, el pH, la conductividad eléctrica, la demanda química de oxígeno (DQO), la turbidez, los nutrientes y otros compuestos permiten identificar procesos de contaminación, degradación ambiental y cambios en la dinámica ecológica del sistema hídrico.

Desde la perspectiva de la Ciencia de Datos, esta información representa una fuente de variables explicativas de alta relevancia para el análisis multivariado, la detección de patrones espacio-temporales y el entrenamiento de modelos predictivos orientados a estimar el comportamiento futuro de la calidad del agua.

La utilización de información oficial proveniente de la plataforma de Datos Abiertos de Colombia, administrada por la Corporación Autónoma Regional (CAR), garantiza la trazabilidad, confiabilidad y representatividad de los registros utilizados en el Framework.

Adicionalmente, la incorporación de procesos de estandarización, validación estructural, auditoría, generación de metadatos y gobierno del dato permite asegurar la interoperabilidad de esta capa con las demás fuentes de información del Framework, facilitando posteriormente la integración con variables climáticas, hidrológicas, institucionales y de percepción ciudadana.

# **Alcance de la capa**

La Capa de Calidad del Agua comprende el procesamiento integral de la información fisicoquímica registrada por la red oficial de monitoreo del río Bogotá, incluyendo:

* Extracción de información desde la plataforma de Datos Abiertos de Colombia.
* Consolidación de registros históricos correspondientes al período 2005–2024.
* Integración de los puntos oficiales de monitoreo ubicados a lo largo de la cuenca del río Bogotá.
* Conservación de la totalidad de los parámetros fisicoquímicos disponibles para preservar el valor científico del conjunto de datos.
* Estandarización de la estructura de datos, tipos de información e identificadores únicos.
* Incorporación de mecanismos de gobierno del dato mediante metadatos, auditoría, control de versiones, trazabilidad e integridad mediante funciones hash.
* Validación de la consistencia estructural, cobertura espacial, cobertura temporal, distribución de parámetros y calidad general de la información.
* Generación de un conjunto de datos interoperable preparado para su integración con las demás capas del Framework.

Como resultado se obtiene el archivo 02_Capa_Calidad_Agua_Framework.xlsx, el cual constituye la fuente oficial de información de calidad del agua dentro del Framework V7.

---

Un aspecto que considero especialmente valioso de esta capa

A diferencia de las capas Climática e Hidrológica, en esta decidimos no reducir el conjunto de datos a los 10 parámetros más frecuentes. Conservamos los 68 parámetros de calidad del agua, incluso aquellos con menor frecuencia de medición o con valores ausentes asociados a límites de detección o campañas específicas. Esa decisión fortalece el Framework porque mantiene el valor científico del conjunto de datos original y deja abierta la posibilidad de que futuras etapas de modelado o análisis seleccionen las variables más adecuadas según cada caso de uso, sin perder información durante la fase de ingeniería de datos. Creo que esa decisión metodológica merece destacarse explícitamente en la tesis, ya que evidencia un enfoque orientado a preservar la integridad de la información antes que simplificarla prematuramente.

## **M3 – Estandarización**

````python
# Crear identificadores

df.insert(
    0,
    "ID_Registro",
    [
        f"{CODIGO_CAPA}-{str(i).zfill(7)}"
        for i in range(1, len(df)+1)
    ]
)

df["ID_Capa"] = CODIGO_CAPA
````

````python
# Ordenar columnas

columnas_control = [
    "ID_Registro",
    "ID_Capa"
]

otras_columnas = [
    c for c in df.columns
    if c not in columnas_control
]

df = df[columnas_control + otras_columnas]
````

````python
# Normalización de tipos

df["resultado"] = pd.to_numeric(
    df["resultado"],
    errors="coerce"
)

df["fecha"] = pd.to_datetime(
    df["fecha"],
    errors="coerce"
)

df["codigo__muestra"] = df["codigo__muestra"].astype(str)

df["codigoestacion"] = (
    df["nombre_del_punto_de_monitoreo"]
    .astype(str)
)
````

¿Por qué crear codigoestacion?

Porque las demás capas poseen un identificador principal del punto de monitoreo.

En Calidad de Agua no existe uno.

El campo más estable es: nombre_del_punto_de_monitoreo

Por eso podemos construir un identificador lógico.

Más adelante incluso podremos reemplazarlo por un código hash si lo deseas.

````python
# Eliminar espacios

df.columns = df.columns.str.strip()

columnas_texto = [
    "nombre_del_punto_de_monitoreo",
    "corriente",
    "zona_hidrogr_fica_zh",
    "nombre_subzona_hidrogr_fica",
    "departamento",
    "municipio",
    "propiedad_observada",
    "unidad_del_resultado",
    "proyecto"
]

for c in columnas_texto:
    df[c] = df[c].astype(str).str.strip()
````

````python
# Ordenar registros

df = df.sort_values(
    by=[
        "nombre_del_punto_de_monitoreo",
        "fecha",
        "propiedad_observada"
    ]
).reset_index(drop=True)
````

## **M4 – Gobierno del Dato**

````python
# Fecha ETL

fecha = datetime.now()

df["Fecha_ETL"] = fecha.strftime("%Y-%m-%d")

df["Hora_ETL"] = fecha.strftime("%H:%M:%S")
````

````python
# Versión Framework

df["Version_Framework"] = VERSION
````

````python
# Estado del registro

df["Estado_Registro"] = "VALIDADO"
````

````python
# Responsable

df["Responsable"] = RESPONSABLE
````

````python
# Hash de Integridad

def generar_hash(fila):

    texto = "".join(fila.astype(str))

    return hashlib.md5(texto.encode()).hexdigest()

df["Hash"] = df.astype(str).apply(generar_hash, axis=1)
````

Observación importante para esta capa

Aquí hay una diferencia importante respecto a las dos primeras capas.

En la Capa Climática y la Capa Hidrológica, cada fila representaba una única observación (clima mensual o nivel hidrológico), por lo que el hash era completamente estable.

En Calidad del Agua, cada fila representa una medición de un parámetro específico (pH, DQO, OD, Conductividad, etc.) realizada en un punto y una fecha determinados. Por ello, el hash se calcula sobre un conjunto de atributos diferente, pero el mecanismo es exactamente el mismo. Esto mantiene la trazabilidad e integridad del dato sin necesidad de modificar el algoritmo.
