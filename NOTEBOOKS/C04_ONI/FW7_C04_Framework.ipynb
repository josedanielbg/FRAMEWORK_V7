# **CAPA ONI**


# **Objetivo de la Capa**

Desarrollar una capa de información climática basada en el Índice Oceánico del Niño (ONI), estandarizada bajo la arquitectura del Framework V7, con el propósito de incorporar la variabilidad climática de gran escala como variable exógena en los modelos de predicción hidrológica. Esta capa permite representar la influencia de los fenómenos El Niño, La Niña y condiciones Neutrales sobre la dinámica hidrológica de la cuenca del río Bogotá, facilitando la integración de señales climáticas globales con las variables hidrológicas, meteorológicas y ambientales utilizadas por el modelo LSTM.

#  **Justificación Científica**

El comportamiento hidrológico de una cuenca no depende exclusivamente de las condiciones locales de precipitación, temperatura o caudal, sino que también está influenciado por procesos climáticos de gran escala asociados a la variabilidad océano-atmósfera. Uno de los principales mecanismos que explica esta interacción es el fenómeno ENSO (El Niño–Oscilación del Sur), cuya intensidad es cuantificada mediante el Índice Oceánico del Niño (ONI), calculado a partir de las anomalías de temperatura superficial del océano Pacífico ecuatorial en la región Niño 3.4.

Diversos estudios han demostrado que las fases cálidas (El Niño) y frías (La Niña) modifican los patrones de precipitación, la disponibilidad hídrica y la ocurrencia de eventos extremos en Colombia, generando efectos directos sobre los niveles de los ríos, los embalses y la calidad del agua. Por esta razón, la incorporación del ONI dentro del Framework V7 permite incluir una variable climática de carácter predictivo que antecede temporalmente muchas de las respuestas hidrológicas observadas.

La estandarización de esta serie temporal bajo principios de gobierno del dato garantiza su interoperabilidad con las demás capas del Framework, permitiendo que el modelo LSTM aprenda relaciones temporales entre las anomalías oceánicas y el comportamiento hidrológico de la cuenca del río Bogotá, fortaleciendo así la capacidad predictiva del sistema frente a escenarios de variabilidad climática.

# **Alcance de la capa**

La capa ONI comprende la adquisición, estandarización, validación y documentación de la serie temporal mensual del Índice Oceánico del Niño (ONI), obtenida a partir de la información oficial publicada por la National Oceanic and Atmospheric Administration (NOAA). El proceso incluye la extracción automática de la información, la normalización de fechas, la clasificación de los estados ENSO (Niño, Niña y Neutral), la incorporación de metadatos, mecanismos de auditoría, indicadores de calidad y elementos de gobierno del dato definidos por el Framework V7.

El alcance de esta capa se limita a representar la señal climática global asociada al fenómeno ENSO y no pretende modelar directamente variables hidrológicas o meteorológicas locales. Su función principal consiste en aportar una variable exógena de carácter climático que será integrada posteriormente con las capas de clima, hidrología, calidad del agua, embalses y demás componentes del Framework para la construcción del conjunto de datos maestro que alimentará el modelo predictivo basado en redes neuronales LSTM.

## **M3 – Estandarización**

````python
# Crear identificadores

df.insert(
    0,
    "ID_Registro",
    [
        f"{CODIGO_CAPA}-{str(i).zfill(7)}"
        for i in range(1, len(df)+1)
    ]
)

df["ID_Capa"] = CODIGO_CAPA
````

````python
# Ordenar columnas

columnas_control = [
    "ID_Registro",
    "ID_Capa"
]

otras_columnas = [
    c
    for c in df.columns
    if c not in columnas_control
]

df = df[columnas_control + otras_columnas]
````

````python
# Normalización de tipos

df["Fecha"] = pd.to_datetime(df["Fecha"])

df["ONI"] = df["ONI"].astype(float)

df["Estado_ENSO"] = df["Estado_ENSO"].astype(str)
````

````python
# Eliminar espacios

df.columns = df.columns.str.strip()

df["Estado_ENSO"] = df["Estado_ENSO"].str.strip()
````

````python
# Ordenar registros

df = (
    df
    .sort_values(by="Fecha")
    .reset_index(drop=True)
)
````

````python
print(df.head())

print()

print(df.columns)

print()

print(df.dtypes)
````

## **M4 – Gobierno del Dato**

````python
# Fecha ETL

fecha = datetime.now()

df["Fecha_ETL"] = fecha.strftime("%Y-%m-%d")

df["Hora_ETL"] = fecha.strftime("%H:%M:%S")
````

````python
# Versión del Framework

df["Version_Framework"] = VERSION
````

````python
# Estado del registro

df["Estado_Registro"] = "VALIDADO"
````

````python
# Responsable

df["Responsable"] = RESPONSABLE
````

````python
# Hash de integridad

def generar_hash(fila):

    texto = "".join(fila.astype(str))

    return hashlib.md5(texto.encode()).hexdigest()


df["Hash"] = (
    df.astype(str)
      .apply(generar_hash, axis=1)
)
````
